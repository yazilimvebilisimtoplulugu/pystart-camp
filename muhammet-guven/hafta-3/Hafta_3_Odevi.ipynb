### Soru 2: Kutuphane Sınıfı ve Temel Menü

`Kutuphane` sınıfını ve menüyü tasarla:

- Kitapları bir `kitaplar` listesinde tutar.
- `kitap_ekle(kitap)`: Listeye kitap ekler.
- `kitap_bul(baslik)`: Başlığa göre kitabı bulur (büyük-küçük harf fark etmez).
- `ara(kelime)`: Başlığında kelime geçen kitapların listesini döndürür.
- `odunc_ver(baslik)` ve `iade_al(baslik)`: Kitabı bulup ilgili `Kitap` metodunu çağırır.

Ayrıca ekleme, listeleme, ödünç verme, iade alma, arama ve çıkış seçenekleri olan bir menü yaz. Menü, çıkış seçilene kadar tekrar gösterilsin.

````python
class Kitap:
    # Kodunuzu buraya yazınız
    def __init__(self,baslik,yazar,yil,odunc_mu = False):
        self.baslik = baslik
        self.yazar = yazar
        self.yil = yil
        try:
          if not isinstance(yil,int):
               raise ValueError("Sayısal değer girmelisiniz")
        except ValueError as ex:
              print(ex)
              return 
        self.odunc_mu = odunc_mu
        self.iade_et()  # bunu yazma sebebim kitap raftamı sorusuna eğer ben iade veya ödünç fonksiyonuna sokmazsam False yazıyordu ondan burdanda bir geçsin fonksiyona girmeyen kitaplar dedim yani işlem yoksa rafta yazar.
    def odunc_al(self):
            self.odunc_mu = True
            self.odunc_mu = "Ödünç"
    def iade_et(self):
            self.odunc_mu = False
            self.odunc_mu = "Rafta"
            return self.odunc_mu
    def __str__(self):
        return f"{self.baslik} - {self.yazar} ({self.yil}) [{self.odunc_mu}]"


class Kutuphane():
    kitaplistesi = []
    kitaplistesi_sadece_Selfli = [] # burayı ekleme sebebim araştırdım ve self.kitap şeklinde ekleyince sadece isim gidiyordu ödünç verme iade alma durumunda işime yaramazdı bende böyleli yaptım bi tane
    def __init__(self,kitap,odunc_mu=False):
        self.kitap = kitap
        self.odunc_mu = odunc_mu
        self.kitaplistesi.append(self.kitap)
        self.kitaplistesi_sadece_Selfli.append(self)
    def odunc_durumu(self,baslik):
        self.baslik = baslik
        if self.odunc_mu == True:
                raise KitapMusaitDegil("Bu kitap zaten ödünçte müsait değil.")
        else:
                self.odunc_mu = True
                print(f"{self.kitap} adlı kitap başarıyla ödünç verildi.")
    def iade_durumu(self,baslik):
        self.baslik = baslik
        if self.odunc_mu == True:
            print(f"{self.kitap} adlı kitabın iadesi alınıp rafa eklendi")
            self.odunc_mu = False
        else:
            self.odunc_mu = False
            raise KitapMusaitDegil("Bu kitap zaten rafta")
    def kaydet(self):
        kitap_classına_at = input("kaydedilecek kitabın başlık,yazar,yıl formatında özelliklerini yaz")
        self.kitap = Kitap(kitap_classına_at)
    def yukle():
        try:
            with open("kitaplar.txt","r",encoding="utf-8") as file:
                for i in file:
                 print(i)
        except FileNotFoundError:
            print("Rafta kitap yok")

        
                






def kitap_ekle(kitap):
    kitap = Kutuphane(kitap)
    print(f"{kitap.kitap} adlı kitap başarıyla eklendi")

def kitap_bul(baslik=str):
    baslik.lower()
    for x in range(len(Kutuphane.kitaplistesi)):
        if baslik == Kutuphane.kitaplistesi[x]:
            print(f"{baslik} adlı aradığınız kitap bulundu")
            break
    else:
        raise KitapBulunamadi("Aradığınız kitap kütüphanede mevcut değil")



def ara(kelime=str):
    kelime.lower()
    i = 0
    for x in range(len(Kutuphane.kitaplistesi)):
        if kelime in Kutuphane.kitaplistesi[x]:
            i +=1
            print(f"{i}. {Kutuphane.kitaplistesi[x]}")
        elif kelime not in Kutuphane.kitaplistesi[x]:
            raise KitapBulunamadi("Aradığınız kelimeyi içeren kitaplar bulunamadı")




def odunc_ver(baslik):
    for k in Kutuphane.kitaplistesi_sadece_Selfli:
        if k.kitap == baslik:
            k.odunc_durumu(baslik)
            break
    else:
        raise KitapBulunamadi("Bu kitap kütüphanede bulunmuyor") or KitapMusaitDegil("Bu kitap zaten ödünçte müsait değil.")
                
def iade_al(baslik):
    for k in Kutuphane.kitaplistesi_sadece_Selfli:
       if k.kitap == baslik:
           k.iade_durumu(baslik)
           break
       else:
           raise KitapMusaitDegil("Bu kitap zaten rafta")


class KitapBulunamadi(Exception):
    pass

class KitapMusaitDegil(Exception):
    pass



while True:
    menu = input(f"{"*" * 15} Kütüphane Menüsü {"*" * 15}\nkitap eklemek için : kitap ekle  || kitapları görmek için : kitap listesi \nKitap ödünç vermek için : ödünç ver || kitap iadesi almak için : iade al\nKitabı isme göre aramak için : kitap ara || kelimeye göre aramak için : kelime ara \nKaydetme modu için : Kayıt ||Menüden çıkmak için q basınız :")
    if menu == "q":
        break
    elif menu == "kitap ekle":
        kitap_isim = input("kitap ismini giriniz : ")
        kitap_ekle(kitap_isim)
    elif menu == "kitap listesi":
        print(Kutuphane.kitaplistesi)
    elif menu == "ödünç ver":
        ödünc_isim = input("Ödünç verilecek kitap ismini giriniz : ")
        try:
            odunc_ver(ödünc_isim)
        except KitapBulunamadi as ex:
            print(ex)
        except KitapMusaitDegil as müsaitlik:
            print(müsaitlik)

    elif menu == "iade al":
        iade_isim = input("iade alınacak kitabın ismini giriniz : ")
        try:
            iade_al(iade_isim)
        except KitapMusaitDegil as ex:
            print(ex)

    elif menu == "kitap ara":
        kitap_ara = input("Aranacak kitabın ismini giriniz : ")
        try:
            kitap_bul(kitap_ara)
        except KitapBulunamadi as ex:
            print(ex)
    elif menu == "kelime ara":
        kelime = input("Arancak kelimeyi giriniz : ")
        try:
            ara(kelime)
        except KitapBulunamadi as ex:
            print(ex)
    elif menu == "Kayıt":
        Kutuphane.kaydet()
    else:
        print("Geçersiz komut")
````

### Soru 5: Iterator, Generator ve Kendi Modülün

- Raftaki kitapları tek tek `yield` eden bir `musait_kitaplar()` generator metodu yaz.
- Menüye raftaki kitapları listeleyen yeni bir seçenek (`6) Raftakiler`) ekle.
- **Kendi modülün:** Hata sınıflarının, `Kitap` ve `Kutuphane` sınıflarının son halini aşağıdaki `%%writefile kutuphane.py` hücresine yaz. Hücre çalışınca kod `kutuphane.py` dosyasına kaydedilir. Menü hücresi sınıfları `from kutuphane import ...` ile alsın.

**Menü Seçenekleri:**
1) Kitap ekle | 2) Listele | 3) Ödünç ver | 4) İade al | 5) Ara | 6) Raftakiler | 0) Kaydet ve çık
