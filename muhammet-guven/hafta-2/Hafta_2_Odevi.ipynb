### Soru 1: Kargo Ücreti

Kullanıcıdan paketin ağırlığını (kg), gönderim mesafesini (km) ve hızlı kargo isteyip istemediğini (`e/h`) al. Ücreti şu tarifeye göre hesapla:

| Ağırlık | Ücret |
| :--- | :--- |
| 2 kg'a kadar (2 dahil) | 50 TL |
| 2-5 kg (5 dahil) | 80 TL |
| 5-10 kg (10 dahil) | 120 TL |
| 10 kg üstü | 120 TL + 10 kg'ı aşan her kg için 10 TL |

- Mesafe 500 km'den fazlaysa ücrete **%20** ekle.
- Hızlı kargo seçildiyse ücreti **1.5** ile çarp.
- Ağırlık ya da mesafe 0 veya negatifse ücret hesaplama, `'Geçersiz ağırlık ya da mesafe.'` yaz.

**Örnek Çıktı:**
```text
Paketin ağırlığı (kg): 12.5
Mesafe (km): 600
Hızlı kargo olsun mu? (e/h): e
Kargo ücreti: 261.00 TL
```

> **Kontrol:** Kodunu sınır değerlerle de dene: 2 kg 50 TL, 2.1 kg 80 TL tutmalı.

````python
# --- Soru 1 Kod Alanı ---
# Kodunuzu buraya yazınız.
paketkg = float(input(("Paket kaç kilo giriniz : ")))
km = float(input("Gönderim mesafesini giriniz : "))
hizlikargo = input("Hızlı kargo olsun mu ? yalnızca evet yada hayır yazınız : ")
fiyat = 0
if (paketkg > 0) and (paketkg <= 2):
    fiyat += 50
elif (paketkg > 2) and (paketkg <= 5):
    fiyat += 80
elif (paketkg > 5) and (paketkg < 10):
    fiyat += 120
elif paketkg > 10:
    fiyat = 120 + ((paketkg - 10) * 10)
    if km > 500 and hizlikargo == "evet":
        fiyat = fiyat + fiyat * 0.2
        fiyat = fiyat * 1.5
    elif km > 500:
        fiyat = fiyat + fiyat * 0.2
    elif hizlikargo == "evet":
        fiyat = fiyat * 1.5
if paketkg <= 0:
    print("Hatalı mesafe veya kilo.")
else:
    print(f"Paketin ağırlığı : {paketkg:.2f}\nMesafe (km) : {km:.2f}\nHızlı kargo olsun mu? : {hizlikargo}\nKargo ücreti : {fiyat:.2f}")
````

### Soru 4: Fonksiyonlar

Aşağıdaki fonksiyonları yaz ve her birini en az bir kez çağırıp sonucunu yazdır. Fonksiyonlar sonucu ekrana yazmasın, `return` ile döndürsün.

1. `sesli_harf_say(metin)`: Metindeki sesli harf sayısını döndürür. `sesli_harf_say('Python öğreniyorum')` sonucu `6` olmalı.
2. `en_uzun_kelime(cumle)`: Cümledeki en uzun kelimeyi döndürür; eşitlikte ilk geleni seçer. `'Kod yazmak pratik ister'` için sonuç `'yazmak'`.
3. `istatistik(*sayilar)`: İstediğin kadar sayı alır; en küçük, en büyük ve ortalamayı birlikte döndürür. `istatistik(4, 8, 15, 16, 23, 42)` sonucu `(4, 42, 18.0)`.
4. `selamla(isim, dil='tr')`: Dil 'tr' ise 'Merhaba, Ayşe!', 'en' ise 'Hello, Ayşe!' döndürür. Dil verilmezse Türkçe selamlar.
5. Fonksiyon yazmadan: `['kod', 'python', 'döngü', 'if', 'fonksiyon', 'liste']` listesinden `filter()` ve `lambda` ile 4 harften uzun kelimeleri seç, `map()` ve `lambda` ile büyük harfe çevir.

````python
# --- Soru 4 Kod Alanı ---
# Kodunuzu buraya yazınız.
#"1."
def sesli_harf_say(metin):
    metin = str(metin)
    metin = metin.lower()
    a = metin.count("a")
    b = metin.count("e")
    c = metin.count("ı")
    d = metin.count("i")
    e = metin.count("o")
    f = metin.count("ö")
    g = metin.count("u")
    h = metin.count("ü")
    return a+b+c+d+e+f+g+h
    
a = sesli_harf_say("python öğreniyorum")
print(a)

# 2.
def en_uzun_kelime(cumle):
   cumle = str(cumle)
   cumle = cumle.split( )
   return max(cumle, key=len)

a = en_uzun_kelime("kod yazmak pratik ister")
print(a)


# 3.
def istatisik(*sayilar):
    return min(sayilar), max(sayilar), sum(sayilar) / len(sayilar)

k = istatisik(4, 8, 15, 16, 23, 42)
print(k)

# 4.
def selamla(isim,dil = "tr" or "en"):
    if dil == "":
        return f"Merhaba {isim}"
    if dil == "en":
        return f"Hello {isim}"
    elif dil == "tr":
        return f"Merhaba {isim}"

slm = selamla("Muhammet")
print(slm)

#5.
liste = ["kod","python","döngü","if","fonksiyon","liste"]
listefor = [x for x in liste]
dortden_buyukler = (list(filter(lambda a : len(a) > 4,listefor)))
buyuk_harfli_hali = list(map(lambda a: a.upper(),dortden_buyukler))
print(buyuk_harfli_hali)
````

### Soru 5: Adam Asmaca

Bilgisayarın tuttuğu kelimeyi harf harf tahmin ettiğin bir oyun yaz.

- Program en az 10 kelimelik bir listeden rastgele bir kelime seçsin: `kelime = random.choice(kelimeler)`. `random` modülünü 3. haftada göreceksin; şimdilik bu satırı olduğu gibi kullanabilirsin.
- Kelime başta `_ _ _ _` olarak gösterilsin; doğru tahmin edilen harfler yerlerine yazılsın.
- Oyuncunun 6 yanlış tahmin hakkı olsun. Her turda kalan hak ve denenen harfler gösterilsin.
- Daha önce denenen bir harf tekrar girilirse uyarı verilsin ve hak düşmesin.
- Tek harf olmayan girişler (boş, birden fazla karakter, rakam) geçersiz sayılsın.
- Kelime tamamlanınca kazandınız, haklar bitince kaybettiniz mesajı çıksın ve kelime gösterilsin.
- Kod en az iki fonksiyondan oluşsun.

**Örnek Çıktı:**
```text
Adam Asmaca'ya hoş geldin!

_ _ _ _ _
Kalan hak: 6 | Denenen harfler:
Bir harf tahmin et: a
Yanlış!

_ _ _ _ _
Kalan hak: 5 | Denenen harfler: a
Bir harf tahmin et: ö
Doğru!

_ ö _ _ _
Kalan hak: 5 | Denenen harfler: a ö
Bir harf tahmin et: ö
'ö' harfini zaten denedin.
```

````python
import random

# --- Soru 5 Kod Alanı ---
# Kodunuzu buraya yazınız.
kelimeler = ["elma","armut","python","java","düzce","kodlama","algoritma","bilgisayar","sonbahar","kış"] #burda kafama göre kelimeler yazdım
kelime = random.choice(kelimeler)
kelime_sansurlu_hali = len(kelime) * "_" #kelime kaç harfliyse ona göre sansürlesin dedim
hak = 6 # burda hak sayısını önceden belirledim çünkü fonksiyonlarda kullanacağım

def harf_girisi(girilen_harf):
    global hak
    global kelime
    global kelime_sansurlu_hali
    if girilen_harf in kelime:
        for x in range(len(kelime)):
            if kelime[x] == girilen_harf:
                kelime_sansurlu_hali_listesi = list(kelime_sansurlu_hali)
                kelime_sansurlu_hali_listesi[x] = girilen_harf
                kelime_sansurlu_hali = "".join(kelime_sansurlu_hali_listesi) # burdaki mantıksa ben bu harflerin yerini değiştireyim fakat str olduğu için hata veriyodu bende internetten böyle bir taktik buldum önce list yaptık sonra indeks numarasıyla yer değiştirip tekrar stringe çevirdik.
                continue
        print("Girdiğiniz harf doğru")
    else:
            hak -=1
            print("Girdiğiniz kelime yanlış 1 hakkınız gitti.")


def ihtimaller(girilen_harf):   #burda bazı ihtimalleri yazarım dedim ama hepsini buraya koyamayınca bazısını fonksiyonsuz yazdım bir de elif te ki durumda çalışmıyor
    girilen_harf = str(girilen_harf)
    if len(girilen_harf) >1:
        print("Yalnızca harf girebilirsiniz hakkınızdan gitmedi")
    elif girilen_harf is int:
        print("Sayı giremezsiniz yalnızca harf girmelisiniz")


print("Adam asmacaya hoş geldiniz!")
girilen_harfler = "" #girilen harfleri bi kenarda toplamak için önceden belirledim
while hak != 0:
    girilen_harf = input("tahmin etmek için harf giriniz : ")
    if girilen_harf in girilen_harfler:
        print("Bu harfi zaten girdiniz hakkınızdan gitmedi")
        continue
    elif len(girilen_harf) > 1:
        print("Yalnızca harf girebilirsiniz hakkınızdan gitmedi")
        continue
    girilen_harfler = girilen_harfler + " " + girilen_harf
    ihtimaller(girilen_harf)
    harf_girisi(girilen_harf)
    if hak == 0:
        print(f"Hakkınız bitti doğru kelime {kelime} idi")
        break
    elif kelime_sansurlu_hali == kelime:
        print(f"Tebrikler kazandınız kelime {kelime} idi")
        break 
    print(f"{kelime_sansurlu_hali}\n Kalan hak : {hak} | Denenen harfler : {girilen_harfler}")

# bu soruyu yaparken internetten hem araştırma yaptım hemde yapay zekaya bazı hataları sordum (kod yazdırmadım ona) çünkü baya bi hatalı şeyler yaptım ve hepsini sadece sitelerde aramaya kalksam bitmezdi belkide.
# bu soru bana karmaşık geldiği için böyle bir açıklama ihtiyacı duydum

````
