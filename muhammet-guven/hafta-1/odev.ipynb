### Soru 3: Not Listesi

Kullanıcıdan üç notu tek satırda, aralarında boşlukla al (`70 85 90`). Notları sayıya çevirip bir listeye koy, ardından:

1. Kullanıcıdan bir not daha alıp listeye ekle.
2. Listeyi küçükten büyüğe sırala ve yazdır.
3. En düşük notu, en yüksek notu ve ortalamayı yazdır.
4. En düşük notu listeden çıkar ve yeni ortalamayı yazdır.

> **İpucu:** `split()` metni boşluklardan böler ama parçalar hâlâ metindir.

**Örnek Çıktı:**
```text
3 notu aralarında boşluk bırakarak yaz: 70 85 90
Bir not daha ekle: 60
Notlar (sıralı): [60, 70, 85, 90]
En düşük: 60 | En yüksek: 90
Ortalama: 76.25
En düşük not çıkarıldı: [70, 85, 90]
Yeni ortalama: 81.67
```

````python
# --- Soru 3 Kod Alanı ---
# Kodunuzu buraya yazınız.
notlar = input("boşluk bırakarak 3 tane not giriniz : ")
notlar = notlar.split( )
notlar[0],notlar[1],notlar[2] = (int(notlar[0]),int(notlar[1]),int(notlar[2]))
not2 = int(input("Bir diğer notu giriniz : "))
notlar.append(not2)
notlar.sort()
print(f"Sıralı hali : {notlar}")
dusuknot = min(notlar)
yukseknot = max(notlar)
ortalama = (notlar[0] + notlar[1] + notlar[2] + notlar[3]) / len(notlar)
print(f"En düşük not : {dusuknot} || En yüksek not : {yukseknot}\nOrtalama : {float(ortalama):.2f}")
notlar.remove(min(notlar))
print(f"En düşük not çıkarılmış hali : {notlar}")
yeniortalama = (notlar[0] + notlar[1] + notlar[2]) / len(notlar)
print(f"Yeni ortalama {float(yeniortalama):.2f}")

````

### Soru 5: Ortak Hobiler

İki kişinin hobilerini virgülle ayrılmış olarak al (`futbol,satranç,müzik`) ve her birini bir kümeye (`set`) çevir. Şunları yazdır:

1. Ortak hobiler, yalnızca birinci kişide olanlar, yalnızca ikinci kişide olanlar ve tüm hobiler
2. Şu soruların cevabı (`True` ya da `False` olarak):
   - Ortak hobiniz var mı?
   - Hobi sayılarınız eşit mi?
   - İkiniz de en az 3 hobi yazdınız mı?
   - 'satranç' ikinizden birinin hobisi mi?

> **İpucu:** Kümeleri yazdırırken sıra her çalıştırmada değişebilir; `sorted()` ile sabitleyebilirsin.

````python
# --- Soru 5 Kod Alanı ---
# Kodunuzu buraya yazınız.
birincikisi = input("1.kişi hobilerini virgül ile ayırarak girsin : ")
ikincikisi = input("2.kişi hobilerini virgül ile ayırarak girsin : ")
birincikisi = birincikisi.split(",")
ikincikisi = ikincikisi.split(",")
birinciset = set()
ikinciset = set()
birinciset.update(birincikisi)
ikinciset.update(ikincikisi)
ortakhobiler = birinciset.intersection(ikinciset)
print(f"Ortak hobiler : {ortakhobiler}")
print(f"Birinci kişinin ortak olmayan hobileri : {birinciset.difference(ikinciset)}")
print(f"İkinci kişinin ortak olmayan hobileri : {ikinciset.difference(birinciset)}")
print(f"Bütün hobiler : {birinciset.union(ikinciset)}")
print(f"Ortak hobiniz var mı ? : {not birinciset.isdisjoint(ikinciset)}")
print(f"Hobi sayılarınız eşit mi ? : {len(birinciset) == len(ikinciset)}")
print(f"İkinizde en az 3 hobi yazdınız mı ? : {(len(birinciset) >= 3) and (len(ikinciset) >= 3)}")
print(f"Satranç ikinizden birinizin hobisi mi ? : {("satranç" in birinciset) or ("satranç" in ikinciset)}")
# bu soruda yapamadığım tek kısım ortak hobi olmayınca Ortak hobiler : kısmına set() yazması
````
