### Soru 1: Kitap Sınıfı

Aşağıdaki özellik ve metotlara sahip `Kitap` sınıfını yaz:

- **Özellikler:** `baslik`, `yazar`, `yil` ve `odunc_mu` (başta `False`).
- `odunc_ver()`: Kitabı ödünçte olarak işaretler.
- `iade_et()`: Kitabı rafta olarak işaretler.
- `__str__()`: Kitabı tek satırda gösterir: `Suç ve Ceza - Fyodor Dostoyevski (1866) [Rafta]`

> İki kitap oluştur, birini ödünç ver ve ikisini de `print()` ile yazdırarak sınıfını dene.

````python
# --- Soru 1 Kod Alanı ---
#class Kitap:
    # Kodunuzu buraya yazınız
   # pass

# Test: 
class Kitap:
    def __init__(self, baslik, yazar, yil, odunc_mu=False):
        self.baslik = baslik
        self.yazar = yazar
        self.yil = yil
        self.odunc_mu = odunc_mu

    def odunc_ver(self):
        self.odunc_mu = True

    def iade_et(self):
        self.odunc_mu = False

    def __str__(self):
        if self.odunc_mu:
            durum = "Ödünçte"
        else:
            durum = "Rafta"
        return f"{self.baslik} - {self.yazar} ({self.yil}) [{durum}]"


# Test
kitap1 = Kitap("Suç ve Ceza", "Fyodor Dostoyevski", 1866)
kitap2 = Kitap("Sefiller", "Victor Hugo", 1862)

# Birini ödünç verip ikisini de yazdıralım
kitap1.odunc_ver()

print(kitap1)
print(kitap2)
````

````python
### Soru 2: Kutuphane Sınıfı ve Temel Menü

`Kutuphane` sınıfını ve menüyü tasarla:

- Kitapları bir `kitaplar` listesinde tutar.
- `kitap_ekle(kitap)`: Listeye kitap ekler.
- `kitap_bul(baslik)`: Başlığa göre kitabı bulur (büyük-küçük harf fark etmez).
- `ara(kelime)`: Başlığında kelime geçen kitapların listesini döndürür.
- `odunc_ver(baslik)` ve `iade_al(baslik)`: Kitabı bulup ilgili `Kitap` metodunu çağırır.

Ayrıca ekleme, listeleme, ödünç verme, iade alma, arama ve çıkış seçenekleri olan bir menü yaz. Menü, çıkış seçilene kadar tekrar gösterilsin.
````

````python
###### --- Soru 2 Kod Alanı ---
class Kutuphane:
    # Kodunuzu buraya yazınız
    pass

class Kitap:

  def __init__(self, baslik, yazar, yil):
    self.baslik = baslik
    self.yazar = yazar
    self.yil = yil
    self.odunc_alindi_mi = False

  def odunc_ver(self):
    if not self.odunc_alindi_mi:
      self.odunc_alindi_mi = True
      return True
    return False

  def iade_et(self):
    if self.odunc_alindi_mi:
      self.odunc_alindi_mi = False
      return True
    return False


class Kutuphane:

  def __init__(self):
    self.kitaplar = []

  def kitap_ekle(self, kitap):
    self.kitaplar.append(kitap)

  def kitap_bul(self, baslik):
    for k in self.kitaplar:
      if k.baslik.lower() == baslik.lower():
        return k
    return None

  def ara(self, kelime):
    bulunanlar = []
    for k in self.kitaplar:
      if kelime.lower() in k.baslik.lower():
        bulunanlar.append(k)
    return bulunanlar

  def odunc_ver(self, baslik):
    kitap = self.kitap_bul(baslik)
    if kitap:
      if kitap.odunc_ver():
        print(f"'{kitap.baslik}' ödünç verildi.")
      else:
        print("Kitap zaten ödünç verilmiş!")
    else:
      print("Kitap bulunamadı.")

  def iade_al(self, baslik):
    kitap = self.kitap_bul(baslik)
    if kitap:
      if kitap.iade_et():
        print(f"'{kitap.baslik}' iade alındı.")
      else:
        print("Kitap zaten kütüphanede, ödünç alınmamış!")
    else:
      print("Kitap bulunamadı.")


# --- Menü Yapısı ---
kutuphane = Kutuphane()

while True:
  print("\n--- KÜTÜPHANE MENÜSÜ ---")
  print("1. Kitap Ekle")
  print("2. Kitapları Listele")
  print("3. Kitap Ara")
  print("4. Ödünç Ver")
  print("5. İade Al")
  print("6. Çıkış")

  secim = input("Seçiminiz (1-6): ")

  if secim == "1":
    baslik = input("Kitap Adı: ")
    yazar = input("Yazarı: ")
    yil = input("Yayın Yılı: ")
    yeni_kitap = Kitap(baslik, yazar, yil)
    kutuphane.kitap_ekle(yeni_kitap)
    print("Kitap başarıyla eklendi.")

  elif secim == "2":
    if not kutuphane.kitaplar:
      print("Kütüphanede henüz kitap yok.")
    else:
      print("\n--- Kitap Listesi ---")
      for k in kutuphane.kitaplar:
        durum = "Ödünçte" if k.odunc_alindi_mi else "Rafta"
        print(f"- {k.baslik} ({k.yazar}, {k.yil}) -> Durum: {durum}")

  elif secim == "3":
    kelime = input("Aranacak kelime: ")
    sonuclar = kutuphane.ara(kelime)
    if sonuclar:
      print("\nBulunan Kitaplar:")
      for k in sonuclar:
        print(f"- {k.baslik} ({k.yazar})")
    else:
      print("Hiçbir kitap bulunamadı.")

  elif secim == "4":
    baslik = input("Ödünç verilecek kitabın adı: ")
    kutuphane.odunc_ver(baslik)

  elif secim == "5":
    baslik = input("İade edilecek kitabın adı: ")
    kutuphane.iade_al(baslik)

  elif secim == "6":
    print("Programdan çıkılıyor...")
    break
  else:
    print("Geçersiz seçim, tekrar deneyin.")
````

### Soru 3: Hata Yönetimi

Özel hata sınıfları tanımla ve işlemleri güvenli hale getir. Soru 1 ve 2'deki metotları buna göre güncelle.

- Kendi hata sınıflarını yaz: `KitapBulunamadi` ve `KitapMusaitDegil` (`Exception` sınıfından türesin).
- `odunc_ver()` kitap zaten ödünçteyse, `iade_et()` kitap zaten raftaysa `KitapMusaitDegil` fırlatsın (`raise`).
- `kitap_bul()` kitabı bulamazsa `KitapBulunamadi` fırlatsın.
- Menüde bu hataları `try / except` ile yakala ve anlaşılır mesaj göster. Program çökmeksizin devam etsin.
- Yıl sayı olarak girilmezse (`ValueError`) uyarı ver.

````python
# --- Soru 3 Kod Alanı ---

# Özel Hata Sınıfları
class KitapBulunamadi(Exception):
  pass


class KitapMusaitDegil(Exception):
  pass


class Kitap:

  def __init__(self, baslik, yazar, yil):
    self.baslik = baslik
    self.yazar = yazar
    self.yil = int(yil)  # Yıl sayı olmalı
    self.odunc_alindi_mi = False


class Kutuphane:

  def __init__(self):
    self.kitaplar = []

  def kitap_ekle(self, kitap):
    self.kitaplar.append(kitap)

  def kitap_bul(self, baslik):
    for k in self.kitaplar:
      if k.baslik.lower() == baslik.lower():
        return k
    # Kitap bulunamazsa hata fırlatıyoruz
    raise KitapBulunamadi(f"'{baslik}' isimli kitap kütüphanede bulunamadı.")

  def ara(self, kelime):
    bulunanlar = [
        k for k in self.kitaplar if kelime.lower() in k.baslik.lower()
    ]
    return bulunanlar

  def odunc_ver(self, baslik):
    kitap = self.kitap_bul(baslik)  # Bulunamazsa KitapBulunamadi hatası atar
    if kitap.odunc_alindi_mi:
      raise KitapMusaitDegil(f"'{kitap.baslik}' zaten ödünç verilmiş!")
    kitap.odunc_alindi_mi = True
    print(f"'{kitap.baslik}' başarıyla ödünç verildi.")

  def iade_al(self, baslik):
    kitap = self.kitap_bul(baslik)  # Bulunamazsa KitapBulunamadi hatası atar
    if not kitap.odunc_alindi_mi:
      raise KitapMusaitDegil(f"'{kitap.baslik}' zaten rafta, iade edilemez!")
    kitap.odunc_alindi_mi = False
    print(f"'{kitap.baslik}' başarıyla iade alındı.")


# --- Hata Yönetimli Menü Yapısı ---
kutuphane = Kutuphane()

while True:
  print("\n--- KÜTÜPHANE MENÜSÜ (GÜVENLİ) ---")
  print("1. Kitap Ekle")
  print("2. Kitapları Listele")
  print("3. Kitap Ara")
  print("4. Ödünç Ver")
  print("5. İade Al")
  print("6. Çıkış")

  secim = input("Seçiminiz (1-6): ")

  try:
    if secim == "1":
      baslik = input("Kitap Adı: ")
      yazar = input("Yazarı: ")
      yil_girdisi = input("Yayın Yılı (Sayı olarak): ")

      # Yılın sayı olup olmadığını kontrol ediyoruz (ValueError yakalama)
      if not yil_girdisi.isdigit():
        raise ValueError("Yayın yılı geçerli bir sayı olmalıdır!")

      yeni_kitap = Kitap(baslik, yazar, int(yil_girdisi))
      kutuphane.kitap_ekle(yeni_kitap)
      print("Kitap başarıyla eklendi.")

    elif secim == "2":
      if not kutuphane.kitaplar:
        print("Kütüphanede henüz kitap yok.")
      else:
        print("\n--- Kitap Listesi ---")
        for k in kutuphane.kitaplar:
          durum = "Ödünçte" if k.odunc_alindi_mi else "Rafta"
          print(f"- {k.baslik} ({k.yazar}, {k.yil}) -> Durum: {durum}")

    elif secim == "3":
      kelime = input("Aranacak kelime: ")
      sonuclar = kutuphane.ara(kelime)
      if sonuclar:
        print("\nBulunan Kitaplar:")
        for k in sonuclar:
          print(f"- {k.baslik} ({k.yazar})")
      else:
        print("Aramanıza uygun kitap bulunamadı.")

    elif secim == "4":
      baslik = input("Ödünç verilecek kitabın adı: ")
      kutuphane.odunc_ver(baslik)

    elif secim == "5":
      baslik = input("İade edilecek kitabın adı: ")
      kutuphane.iade_al(baslik)

    elif secim == "6":
      print("Programdan çıkılıyor...")
      break
    else:
      print("Geçersiz seçim, lütfen 1-6 arasında bir değer girin.")

  # Hataları programı çökertmeden yakalayan kısım:
  except KitapBulunamadi as e:
    print(f" HATA: {e}")
  except KitapMusaitDegil as e:
    print(f" HATA: {e}")
  except ValueError as e:
    print(f" UYARI: {e}")
  except Exception as e:
    print(f" Beklenmeyen bir hata oluştu: {e}")
````

### Soru 4: Dosya İşlemleri (kitaplar.txt)

Kitapları kalıcı olarak dosyada sakla:

- Her kitap `kitaplar.txt` dosyasında bir satır olur. Alanlar `;` ile ayrılır, aralarında boşluk yoktur; son alan kitap ödünçteyse `1`, raftaysa `0` olur:
  ```text
  Suç ve Ceza;Fyodor Dostoyevski;1866;0
  Simyacı;Paulo Coelho;1988;1
  Kürk Mantolu Madonna;Sabahattin Ali;1943;0
  ```
- `Kutuphane` sınıfına kitapları dosyaya yazan `kaydet()` ve dosyadan okuyan `yukle()` metotlarını ekle.
- Program açılınca dosyadaki kitapları yükler. Dosya yoksa (`FileNotFoundError`) boş başlar.
- Menüdeki çıkış seçeneği `'Kaydet ve çık'` olsun ve kitapları dosyaya yazsın (`encoding='utf-8'`).
- Teslim etmeden önce dosyada en az 3 kitap olsun.

````python
# --- Soru 4 Kod Alanı ---
# Dosya okuma ve yazma metotlarını buraya ekleyiniz.
class Kitap:

  def __init__(self, baslik, yazar, yil, odunc_alindi_mi=False):
    self.baslik = baslik
    self.yazar = yazar
    self.yil = int(yil)
    self.odunc_alindi_mi = bool(odunc_alindi_mi)


class Kutuphane:

  def __init__(self):
    self.kitaplar = []

  def kitap_ekle(self, kitap):
    self.kitaplar.append(kitap)

  def kaydet(self, dosya_adi="kitaplar.txt"):
    """Kitapları dosyaya satır satır kaydeder."""
    with open(dosya_adi, "w", encoding="utf-8") as f:
      for k in self.kitaplar:
        durum = "1" if k.odunc_alindi_mi else "0"
        satir = f"{k.baslik};{k.yazar};{k.yil};{durum}\n"
        f.write(satir)

  def yukle(self, dosya_adi="kitaplar.txt"):
    """Dosyadan kitapları okur ve listeye ekler."""
    try:
      with open(dosya_adi, "r", encoding="utf-8") as f:
        for satir in f:
          satir = satir.strip()
          if satir:  # Boş satırları atla
            parcalar = satir.split(";")
            baslik = parcalar[0]
            yazar = parcalar[1]
            yil = int(parcalar[2])
            odunc_durumu = parcalar[3] == "1"

            kitap = Kitap(baslik, yazar, yil, odunc_durumu)
            self.kitaplar.append(kitap)
      print(f"{len(self.kitaplar)} kitap yüklendi.")
    except FileNotFoundError:
      print(
          f"'{dosya_adi}' bulunamadı, yeni ve boş bir kütüphane başlatılıyor."
      )


# --- Test ve Kullanım ---
kutuphane = Kutuphane()

# Program açılınca dosyadan oku
kutuphane.yukle()

# Eğer dosya ilk kez çalışıyorsa örnek 3 kitabı ekleyebilirsin:
if len(kutuphane.kitaplar) == 0:
  kutuphane.kitap_ekle(Kitap("Suç ve Ceza", "Fyodor Dostoyevski", 1866, False))
  kutuphane.kitap_ekle(Kitap("Simyacı", "Paulo Coelho", 1988, True))
  kutuphane.kitap_ekle(
      Kitap("Kürk Mantolu Madonna", "Sabahattin Ali", 1943, False)
  )

# 'Kaydet ve Çık' seçeneği çağrıldığında çalışacak kod:
kutuphane.kaydet()
print("Kitaplar 'kitaplar.txt' dosyasına kaydedildi.")
````

### Soru 5: Iterator, Generator ve Kendi Modülün

Projenin son halini toparla:

- `Kutuphane` sınıfına `__iter__` metodunu ekle; böylece `for kitap in kutuphane:` yazılabilsin. Listelemeyi bu şekilde yap.
- Raftaki kitapları tek tek `yield` eden bir `musait_kitaplar()` generator metodu yaz.
- Menüye raftaki kitapları listeleyen yeni bir seçenek (`6) Raftakiler`) ekle.
- **Kendi modülün:** Hata sınıflarının, `Kitap` ve `Kutuphane` sınıflarının son halini aşağıdaki `%%writefile kutuphane.py` hücresine yaz. Hücre çalışınca kod `kutuphane.py` dosyasına kaydedilir. Menü hücresi sınıfları `from kutuphane import ...` ile alsın.

**Menü Seçenekleri:**
1) Kitap ekle | 2) Listele | 3) Ödünç ver | 4) İade al | 5) Ara | 6) Raftakiler | 0) Kaydet ve çık

**Örnek Çıktı (menü satırları kısaltıldı):**
```text
3 kitap yüklendi.
Seçimin: 3
Ödünç verilecek kitap: Suç ve Ceza
Ödünç verildi.
Seçimin: 3
Ödünç verilecek kitap: Suç ve Ceza
Hata: 'Suç ve Ceza' zaten ödünç verilmiş.
Seçimin: 6
Kürk Mantolu Madonna - Sabahattin Ali (1943) [Rafta]
Seçimin: 0
Kaydedildi. Görüşmek üzere!
```

> Decorator konusunu (12. modül) 3. oturumda bu proje üzerinde birlikte yazacağız.

````python
%%writefile kutuphane.py
# --- Soru 5: kutuphane.py ---
# Hata sınıflarının, Kitap ve Kutuphane sınıflarının son halini buraya yazınız.
# Bu hücreyi çalıştırınca kod kutuphane.py dosyasına kaydedilir.

# --- Özel Hata Sınıfları ---
class KitapBulunamadi(Exception):
    pass

class KitapMusaitDegil(Exception):
    pass

# --- Kitap Sınıfı ---
class Kitap:
    def __init__(self, baslik, yazar, yil, odunc_alindi_mi=False):
        self.baslik = baslik
        self.yazar = yazar
        self.yil = int(yil)
        self.odunc_alindi_mi = bool(odunc_alindi_mi)

# --- Kütüphane Sınıfı ---
class Kutuphane:
    def __init__(self):
        self.kitaplar = []

    def __iter__(self):
        return iter(self.kitaplar)

    def musait_kitaplar(self):
        for k in self.kitaplar:
            if not k.odunc_alindi_mi:
                yield k

    def kitap_ekle(self, kitap):
        self.kitaplar.append(kitap)

    def kitap_bul(self, baslik):
        for k in self.kitaplar:
            if k.baslik.lower() == baslik.lower():
                return k
        raise KitapBulunamadi(f"'{baslik}' bulunamadı.")

    def ara(self, kelime):
        return [k for k in self.kitaplar if kelime.lower() in k.baslik.lower()]

    def odunc_ver(self, baslik):
        kitap = self.kitap_bul(baslik)
        if kitap.odunc_alindi_mi:
            raise KitapMusaitDegil(f"'{kitap.baslik}' zaten ödünç verilmiş.")
        kitap.odunc_alindi_mi = True
        print("Ödünç verildi.")

    def iade_al(self, baslik):
        kitap = self.kitap_bul(baslik)
        if not kitap.odunc_alindi_mi:
            raise KitapMusaitDegil(f"'{kitap.baslik}' zaten kütüphanede.")
        kitap.odunc_alindi_mi = False
        print("İade alındı.")

    def kaydet(self, dosya_adi="kitaplar.txt"):
        with open(dosya_adi, "w", encoding="utf-8") as f:
            for k in self.kitaplar:
                durum = "1" if k.odunc_alindi_mi else "0"
                f.write(f"{k.baslik};{k.yazar};{k.yil};{durum}\n")

    def yukle(self, dosya_adi="kitaplar.txt"):
        try:
            with open(dosya_adi, "r", encoding="utf-8") as f:
                for satir in f:
                    satir = satir.strip()
                    if satir:
                        parcalar = satir.split(";")
                        baslik, yazar, yil, durum = parcalar[0], parcalar[1], parcalar[2], parcalar[3] == "1"
                        self.kitaplar.append(Kitap(baslik, yazar, yil, durum))
            print(f"{len(self.kitaplar)} kitap yüklendi.")
        except FileNotFoundError:
            print("Dosya bulunamadı, yeni kütüphane oluşturuluyor.")
````

````python
# --- Soru 5: Menü ---
# kutuphane.py değiştiğinde otomatik yeniden yüklensin:
# Soru 5: Menü
# Soru 5: Menü
%load_ext autoreload
%autoreload 2

from kutuphane import Kutuphane, Kitap, KitapBulunamadi, KitapMusaitDegil

kutuphane = Kutuphane()
kutuphane.yukle()

while True:
  print(
      "\n1. Kitap ekle | 2) Listele | 3) Ödünç ver | 4) İade al | 5) Ara | 6)"
      " Raftakiler | 0) Kaydet ve çık"
  )
  secimin = input("Seçimin: ")

  try:
    if secimin == "1":
      baslik = input("Kitap Adı: ")
      yazar = input("Yazarı: ")
      yil = input("Yayın Yılı: ")
      kutuphane.kitap_ekle(Kitap(baslik, yazar, yil))
      print("Kitap eklendi.")

    elif secimin == "2":
      for k in kutuphane:
        durum = "[Ödünçte]" if k.odunc_alindi_mi else "[Rafta]"
        print(f"{k.baslik} - {k.yazar} ({k.yil}) {durum}")

    elif secimin == "3":
      baslik = input("Ödünç verilecek kitap: ")
      kutuphane.odunc_ver(baslik)

    elif secimin == "4":
      baslik = input("İade edilecek kitap: ")
      kutuphane.iade_al(baslik)

    elif secimin == "5":
      kelime = input("Aranacak kelime: ")
      sonuclar = kutuphane.ara(kelime)
      for k in sonuclar:
        print(f"{k.baslik} - {k.yazar}")

    elif secimin == "6":
      print("\n--- Raftaki (Ödünç Alınabilir) Kitaplar ---")
      for k in kutuphane.musait_kitaplar():
        print(f"{k.baslik} - {k.yazar} ({k.yil}) [Rafta]")

    elif secimin == "0":
      kutuphane.kaydet()
      print("Kaydedildi. Görüşmek üzere!")
      break

    else:
      print("Geçersiz seçim!")

  except KitapBulunamadi as e:
    print(f"Hata: {e}")
  except KitapMusaitDegil as e:
    print(f"Hata: {e}")
  except Exception as e:
    print(f"Beklenmeyen hata: {e}")
````
