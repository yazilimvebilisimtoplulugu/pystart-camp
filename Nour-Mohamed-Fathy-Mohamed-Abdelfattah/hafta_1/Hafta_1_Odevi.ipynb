### Soru 1: Katılımcı Kartı

Kullanıcıdan adını, soyadını, bölümünü ve doğum tarihini (gün, ay ve yıl ayrı ayrı) `input()` ile al, aşağıdaki gibi bir kart yazdır.

- Doğum tarihini `(gün, ay, yıl)` şeklinde bir **tuple**'da tut.
- Yaşı `2026 - doğum yılı` olarak hesapla.
- Ad soyadın baş harflerini büyüt, tarihi iki haneli yazdır (`07.03.2006`).

> **İpucu:** `f"{gun:02d}"` sayıyı iki haneye tamamlar.

**Örnek Çıktı Biçimi:**
```text
========================================
PyStart Katılımcı Kartı
========================================
Ad Soyad    : Ayşe Yılmaz
Bölüm       : Bilgisayar Mühendisliği
Doğum       : 07.03.2006
Yaş         : 20
========================================
```

````python
# --- Soru 1 Kod Alanı ---
# Kodunuzu buraya yazınız.
ad = input("Adınızı giriniz: ").strip().title()
soyad = input("Soyadınızı giriniz: ").strip().title()
bolum = input("Bölümünüzü giriniz: ").strip().title()

gun = int(input("Doğum gününüz (1-31): "))
ay = int(input("Doğum ayınız (1-12): "))
yil = int(input("Doğum yılınız (ör: 2006): "))


dogum_tarihi = (gun, ay, yil)

yas = 2026 - dogum_tarihi[2]  


ad_soyad = f"{ad} {soyad}"


print("=" * 40)
print("PyStart Katılımcı Kartı")
print("=" * 40)
print(f"Ad Soyad : {ad_soyad}")
print(f"Bölüm    : {bolum}")
print(f"Doğum    : {dogum_tarihi[0]:02d}.{dogum_tarihi[1]:02d}.{dogum_tarihi[2]}")
print(f"Yaş      : {yas}")
print("=" * 40)
````

### Soru 2: Cümle Analizi

Kullanıcıdan bir cümle al ve şunları yazdır:

1. Karakter sayısı ve boşluklar hariç karakter sayısı
2. Kelime sayısı, ilk kelime ve son kelime
3. Cümlenin tamamen büyük harfli hali ve tersten yazılışı
4. Cümlede kaç tane 'a' harfi olduğu (büyük-küçük fark etmeden)
5. Cümlede 'python' kelimesinin geçip geçmediği (`True` ya da `False`)

> **Gözlem:** İçinde 'i' harfi olan bir cümlenin büyük harfli haline dikkatle bak. Ne fark ettin? Gözlemini yorum satırı olarak yaz.

````python
# --- Soru 2 Kod Alanı ---
# Kodunuzu buraya yazınız.
cumle = input("Bir cümle giriniz: ")


toplam_karakter = len(cumle)
bosluksuz_cumle = cumle.replace(" ", "")
bosluksuz_karakter = len(bosluksuz_cumle)


kelimeler = cumle.split()
kelime_sayisi = len(kelimeler)
ilk_kelime = kelimeler[0]
son_kelime = kelimeler[-1]


buyuk_cumle = cumle.upper()
tersten_cumle = cumle[::-1]


a_sayisi = cumle.lower().count("a")


python_var_mi = "python" in cumle.lower()


print("\n--- CÜMLE ANALİZİ SONUÇLARI ---")
print(f"Toplam karakter sayısı               : {toplam_karakter}")
print(f"Boşluklar hariç karakter sayısı       : {bosluksuz_karakter}")
print(f"Kelime sayısı                         : {kelime_sayisi}")
print(f"İlk kelime                            : {ilk_kelime}")
print(f"Son kelime                            : {son_kelime}")
print(f"Büyük harfli hali                     : {buyuk_cumle}")
print(f"Tersten yazılışı                      : {tersten_cumle}")
print(f"Cümledeki 'a' harfi sayısı            : {a_sayisi}")
print(f"Cümlede 'python' kelimesi var mı?    : {python_var_mi}")


# Gözlem Yorumunuz:
# Pythonunun varsayılan dili ingilizce olduğu için .upper() metodunu kullandığımızda küçük i harfi türkçe İ yerine ingilizce I harfine dönüşüyor.
````

### Soru 3: Not Listesi

Kullanıcıdan üç notu tek satırda, aralarında boşlukla al (`70 85 90`). Notları sayıya çevirip bir listeye koy, ardından:

1. Kullanıcıdan bir not daha alıp listeye ekle.
2. Listeyi küçükten büyüğe sırala ve yazdır.
3. En düşük notu, en yüksek notu ve ortalamayı yazdır.
4. En düşük notu listeden çıkar ve yeni ortalamayı yazdır.

> **İpucu:** `split()` metni boşluklardan böler ama parçalar hâlâ metindir.

**Örnek Çıktı:**
```text
3 notu aralarında boşluk bırakarak yaz: 70 85 90
Bir not daha ekle: 60
Notlar (sıralı): [60, 70, 85, 90]
En düşük: 60 | En yüksek: 90
Ortalama: 76.25
En düşük not çıkarıldı: [70, 85, 90]
Yeni ortalama: 81.67
```

````python
# --- Soru 3 Kod Alanı ---
# Kodunuzu buraya yazınız.
girdi = input("3 notu aralarında boşluk bırakarak yazınız:")

parcalar = girdi.split()

not1 = float(parcalar[0])
not2 = float(parcalar[1])
not3 = float(parcalar[2])

notlar = [not1, not2, not3]

yeni_not = float(input("Bir not daha ekleyiniz:"))
notlar.append(yeni_not)

notlar.sort()
print("Notlar: ", notlar)

en_dusuk = min(notlar)
en_yuksek = max(notlar)
ortalama = sum(notlar) / len(notlar)

print(f"En düşük: {en_dusuk} | En yüksek: {en_yuksek}")
print("Ortalama:", round(ortalama, 2))

notlar.pop(0)

yeni_ortalama = sum(notlar) / len(notlar)

print("En düşük not çıkarıldı:", notlar)

print("Yeni ortalama:" round(yeni_ortalama, 2))
````

### Soru 4: Telefon Rehberi

Şu sözlükle başla:
```python
rehber = {'Ali': '0555 111 22 33', 'Ayşe': '0532 444 55 66'}
```

1. Kullanıcıdan bir isim ve numara alıp rehbere ekle.
2. Ali'nin numarasını `'0555 999 88 77'` olarak güncelle.
3. Ayşe'yi rehberden sil ve silinen numarayı yazdır.
4. Kullanıcıdan bir isim al ve numarasını yazdır. İsim rehberde yoksa program hata vermesin, `'Kayıt bulunamadı'` yazsın.
5. Rehberdeki kişi sayısını ve isimleri alfabetik sırayla yazdır.

> **İpucu:** Sözlüklerin `get()` metoduna ikinci bir varsayılan değer verebilirsin.

````python
rehber = {'Ali': '0555 111 22 33', 'Ayşe': '0532 444 55 66'}

# --- Soru 4 Kod Alanı ---
# Kodunuzu buraya yazınız.
rehber = {'Ali': '0555 111 22 33', 'Ayşe': '0532 444 55 66'}

yeni_isim = input("Eklenecek kişinin adı:")
yeni_numara = input("Eklenecek kişinin numarası:")
rehber[yeni_isim] = yeni_numara

rehber['Ali'] = '0555 999 88 77'

silinen_numara = rehber.pop('Ayşe')
print("Ayşenin silinen numarası:" silinen_numara)

aranan_isim = input("Numarasını öğrenmek istediğiniz isim: ")
sonuc = rehber.get(aranan_isim, 'Kayıt bulunamadı')
print(f"{aranan_isim} numarası:", sonuc)

kisi_sayisi = len(rehber)
isimler_alfabetik = sorted(rehber.keys())

print("Rehberdeki toplam kişi sayısı:", kisi_sayisi)
print("Alfabetik kişi listesi:", isimler_alfabetik)
````

### Soru 5: Ortak Hobiler

İki kişinin hobilerini virgülle ayrılmış olarak al (`futbol,satranç,müzik`) ve her birini bir kümeye (`set`) çevir. Şunları yazdır:

1. Ortak hobiler, yalnızca birinci kişide olanlar, yalnızca ikinci kişide olanlar ve tüm hobiler
2. Şu soruların cevabı (`True` ya da `False` olarak):
   - Ortak hobiniz var mı?
   - Hobi sayılarınız eşit mi?
   - İkiniz de en az 3 hobi yazdınız mı?
   - 'satranç' ikinizden birinin hobisi mi?

> **İpucu:** Kümeleri yazdırırken sıra her çalıştırmada değişebilir; `sorted()` ile sabitleyebilirsin.

````python
# --- Soru 5 Kod Alanı ---
# Kodunuzu buraya yazınız.
girdi1 = input("1. Kişinin hobileri (virgülle ayırın): ")
girdi2 = input("2. Kişinin hobileri (virgülle ayırın): ")

kisi1 = set(girdi1.split(","))
kisi2 = set(girdi2.split(","))

ortaklar = kisi1 & kisi2

sadece_kisi1 = kisi1 - kisi2
sadece_kisi2 = kisi2 - kisi1

tum_hobiler = kisi1 | kisi2

print("Ortak hobiler:", sorted(ortaklar))
print("Yalnızca 1. kişide olanlar:", sorted(sadece_kisi1))
print("Yalnızca 2. kişide olanlar:", sorted(sadece_kisi2))
print("Tüm hobiler:", sorted(tum_hobiler))

ortak_var_mi = len(ortaklar) > 0

sayilar_esit_mi = len(kisi1) == len(kisi2)

en_az_3_mu = (len(kisi1) >= 3) and (len(kisi2) >= 3)

satranc_var_mi = "satranç" in tum_hobiler

print("\n--- SONUÇLAR ---")
print("Ortak hobiniz var mı?:", ortak_var_mi)
print("Hobi sayılarınız eşit mi?:", sayilar_esit_mi)
print("İkiniz de en az 3 hobi yazdınız mı?:", en_az_3_mu)
print("'Satranç' ikinizden birinin hobisi mi?:", satranc_var_mi)
````
