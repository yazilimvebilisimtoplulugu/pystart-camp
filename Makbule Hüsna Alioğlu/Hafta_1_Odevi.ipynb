### Soru 1: Katılımcı Kartı

Kullanıcıdan adını, soyadını, bölümünü ve doğum tarihini (gün, ay ve yıl ayrı ayrı) `input()` ile al, aşağıdaki gibi bir kart yazdır.

- Doğum tarihini `(gün, ay, yıl)` şeklinde bir **tuple**'da tut.
- Yaşı `2026 - doğum yılı` olarak hesapla.
- Ad soyadın baş harflerini büyüt, tarihi iki haneli yazdır (`07.03.2006`).

> **İpucu:** `f"{gun:02d}"` sayıyı iki haneye tamamlar.

**Örnek Çıktı Biçimi:**
```text
========================================
PyStart Katılımcı Kartı
========================================
Ad Soyad    : Ayşe Yılmaz
Bölüm       : Bilgisayar Mühendisliği
Doğum       : 07.03.2006
Yaş         : 20
========================================
```

````python
ad = input("Adinizi giriniz: ") 
soyad = input("Soyadinizi giriniz: ")
bolum = input("Bolumunuzu giriniz: ") 

gun = int(input("Dogum gununuzu giriniz (orn: 7): "))
ay = int(input("Dogum ayinizi giriniz (orn: 3): ")) 
yil = int(input("Dogum yilinizi giriniz (orn: 2006): "))

dogum_tarihi = (gun, ay, yil)

yas = 2026 - dogum_tarihi[2]

ad_soyad = f"{ad} {soyad}".title()

formatli_tarih = f"{dogum_tarihi[0]:02d}.{dogum_tarihi[1]:02d}.{dogum_tarihi[2]}"

print("========================================")
print("PyStart Katilimci Karti")
print("========================================")
print(f"Ad Soyad    : {ad_soyad}")
print(f"Bolum       : {bolum}")
print(f"Dogum       : {formatli_tarih}")
print(f"Yas         : {yas}")
print("========================================") 
````

### Soru 2: Cümle Analizi

Kullanıcıdan bir cümle al ve şunları yazdır:

1. Karakter sayısı ve boşluklar hariç karakter sayısı
2. Kelime sayısı, ilk kelime ve son kelime
3. Cümlenin tamamen büyük harfli hali ve tersten yazılışı
4. Cümlede kaç tane 'a' harfi olduğu (büyük-küçük fark etmeden)
5. Cümlede 'python' kelimesinin geçip geçmediği (`True` ya da `False`)

> **Gözlem:** İçinde 'i' harfi olan bir cümlenin büyük harfli haline dikkatle bak. Ne fark ettin? Gözlemini yorum satırı olarak yaz.

````python
cumle = input("Lutfen bir cumle giriniz: ")

toplam_karakter = len(cumle)

bosluksuz_karakter = len(cumle.replace(" ", ""))

kelimeler = cumle.split()
kelime_sayisi = len(kelimeler)

kelimeler = cumle.split() 
ilk_kelime = (kelimeler or ["Yok"])[0]
son_kelime = (kelimeler or ["Yok"])[-1] 

# cumle buyuk harfe gectiginde ingilizce karakterler degisir.
buyuk_harfli = cumle.upper()
ters_yazi = cumle[::-1]  

a_sayisi = cumle.lower().count('a')

python_var_mi = "python" in cumle.lower()

print("\n--- CUMLE ANALIZI ---")
print(f"1. Toplam Karakter Sayisi      : {toplam_karakter}")
print(f"   Bosluksuz Karakter Sayisi   : {bosluksuz_karakter}")
print(f"2. Kelime Sayisi              : {kelime_sayisi}")
print(f"   Ilk Kelime                  : {ilk_kelime}")
print(f"   Son Kelime                  : {son_kelime}")
print(f"3. Buyuk Harfli Hali          : {buyuk_harfli}")
print(f"   Tersten Yazilisi           : {ters_yazi}")
print(f"4. 'a' Harfi Sayisi           : {a_sayisi}")
print(f"5. 'python' Kelimesi Geciyor mu?: {python_var_mi}") 
````

### Soru 3: Not Listesi

Kullanıcıdan üç notu tek satırda, aralarında boşlukla al (`70 85 90`). Notları sayıya çevirip bir listeye koy, ardından:

1. Kullanıcıdan bir not daha alıp listeye ekle.
2. Listeyi küçükten büyüğe sırala ve yazdır.
3. En düşük notu, en yüksek notu ve ortalamayı yazdır.
4. En düşük notu listeden çıkar ve yeni ortalamayı yazdır.

> **İpucu:** `split()` metni boşluklardan böler ama parçalar hâlâ metindir.

**Örnek Çıktı:**
```text
3 notu aralarında boşluk bırakarak yaz: 70 85 90
Bir not daha ekle: 60
Notlar (sıralı): [60, 70, 85, 90]
En düşük: 60 | En yüksek: 90
Ortalama: 76.25
En düşük not çıkarıldı: [70, 85, 90]
Yeni ortalama: 81.67
```

````python
notlar_girdisi = input("3 notu aralarinda bosluk birakarak yaz (orn: 70 85 90): ")

metin_listesi = notlar_girdisi.split()

notlar = [int(notlar_girdisi.split()[0]), int(notlar_girdisi.split()[1]), int(notlar_girdisi.split()[2])] 

yeni_not = int(input("Bir not daha ekle: "))
notlar.append(yeni_not)

notlar.sort()
print(f"Notlar (sirali): {notlar}")

en_dusuk = min(notlar)
en_yuksek = max(notlar)
ortalama = sum(notlar) / len(notlar)

print(f"En dusuk: {en_dusuk} | En yuksek: {en_yuksek}")
print(f"Ortalama: {ortalama:.2f}")

notlar.remove(en_dusuk) 

yeni_ortalama = sum(notlar) / len(notlar)

print(f"En dusuk not cikarildi: {notlar}")
print(f"Yeni ortalama: {yeni_ortalama:.2f}") 
````

### Soru 4: Telefon Rehberi

Şu sözlükle başla:
```python
rehber = {'Ali': '0555 111 22 33', 'Ayşe': '0532 444 55 66'}
```

1. Kullanıcıdan bir isim ve numara alıp rehbere ekle.
2. Ali'nin numarasını `'0555 999 88 77'` olarak güncelle.
3. Ayşe'yi rehberden sil ve silinen numarayı yazdır.
4. Kullanıcıdan bir isim al ve numarasını yazdır. İsim rehberde yoksa program hata vermesin, `'Kayıt bulunamadı'` yazsın.
5. Rehberdeki kişi sayısını ve isimleri alfabetik sırayla yazdır.

> **İpucu:** Sözlüklerin `get()` metoduna ikinci bir varsayılan değer verebilirsin.

````python
rehber = {'Ali': '0555 111 22 33', 'Ayse': '0532 444 55 66'}

yeni_isim = input("Eklenecek kisinin adini giriniz: ")
yeni_numara = input("Eklenecek kisinin numarasini giriniz: ")
rehber[yeni_isim] = yeni_numara

rehber['Ali'] = '0555 999 88 77'

silinen_numara = rehber.pop('Ayse', 'Kayit bulunamadi')
print(f"Ayse rehberden silindi. Silinen Numara: {silinen_numara}") 

aranan_isim = input("Numarasini ogrenmek istediginiz kisinin adini giriniz: ")

numara = rehber.get(aranan_isim, 'Kayit bulunamadi')
print(f"{aranan_isim} - Numara: {numara}")

kisi_sayisi = len(rehber)

alfabetik_isimler = sorted(rehber.keys())

print(f"\n--- GUNCEL REHBER ---")
print(f"Toplam Kisi Sayisi: {kisi_sayisi}")
print(f"Alfabetik isim Listesi: {alfabetik_isimler}")
print(f"Son Rehber Sozlugu: {rehber}") 
````

### Soru 5: Ortak Hobiler

İki kişinin hobilerini virgülle ayrılmış olarak al (`futbol,satranç,müzik`) ve her birini bir kümeye (`set`) çevir. Şunları yazdır:

1. Ortak hobiler, yalnızca birinci kişide olanlar, yalnızca ikinci kişide olanlar ve tüm hobiler
2. Şu soruların cevabı (`True` ya da `False` olarak):
   - Ortak hobiniz var mı?
   - Hobi sayılarınız eşit mi?
   - İkiniz de en az 3 hobi yazdınız mı?
   - 'satranç' ikinizden birinin hobisi mi?

> **İpucu:** Kümeleri yazdırırken sıra her çalıştırmada değişebilir; `sorted()` ile sabitleyebilirsin.

````python
kisi1_girdi = input("1. kisinin hobilerini virgulle ayirarak gir (orn: futbol,satranc,muzik): ")
kisi2_girdi = input("2. kisinin hobilerini virgulle ayirarak gir (orn: sinema,satranc,yuzme,muzik): ")

hobi_kumesi1 = {hobi.strip() for hobi in kisi1_girdi.split(",")}
hobi_kumesi2 = {hobi.strip() for hobi in kisi2_girdi.split(",")}

ortak_hobiler = sorted(hobi_kumesi1.intersection(hobi_kumesi2)) 
sadece_birinci = sorted(hobi_kumesi1.difference(hobi_kumesi2))     
sadece_ikinci = sorted(hobi_kumesi2.difference(hobi_kumesi1))     
tum_hobiler = sorted(hobi_kumesi1.union(hobi_kumesi2))           

ortak_var_mi = len(hobi_kumesi1.intersection(hobi_kumesi2)) > 0

sayilar_esit_mi = len(hobi_kumesi1) == len(hobi_kumesi2)

en_az_3_mu = len(hobi_kumesi1) >= 3 and len(hobi_kumesi2) >= 3

satranc_var_mi = 'satranc' in hobi_kumesi1 or 'satranc' in hobi_kumesi2

print("\n--- HOBI ANALIZI ---")
print(f"1. Ortak Hobiler            : {ortak_hobiler}")
print(f"   Sadece 1. Kiside Olanlar : {sadece_birinci}")
print(f"   Sadece 2. Kiside Olanlar : {sadece_ikinci}")
print(f"   Tum Hobiler (Birlesim)   : {tum_hobiler}")

print("\n2. Sorularin Cevaplari:")
print(f"   - Ortak hobiniz var mi?          : {ortak_var_mi}")
print(f"   - Hobi sayilariniz esit mi?      : {sayilar_esit_mi}")
print(f"   - Ikiniz de en az 3 hobi yazdiniz mi?: {en_az_3_mu}")
print(f"   - 'satranc' birinizin hobisi mi? : {satranc_var_mi}") 
````
