### Soru 1: Kitap Sınıfı

Aşağıdaki özellik ve metotlara sahip `Kitap` sınıfını yaz:

- **Özellikler:** `baslik`, `yazar`, `yil` ve `odunc_mu` (başta `False`).
- `odunc_ver()`: Kitabı ödünçte olarak işaretler.
- `iade_et()`: Kitabı rafta olarak işaretler.
- `__str__()`: Kitabı tek satırda gösterir: `Suç ve Ceza - Fyodor Dostoyevski (1866) [Rafta]`

> İki kitap oluştur, birini ödünç ver ve ikisini de `print()` ile yazdırarak sınıfını dene.

````python
# --- Soru 1 Kod Alanı ---
class Kitap:
    # Kodunuzu buraya yazınız
    def __int__(self, baslik, yazar, yil):
        self.baslik = baslik
        self.yazar = yazar
        self.yil = yil
        self.odunc_mu = False #Başlangıçta kitap alınmamış rafta duruyor.

        def odunc_ver(self):
            self.odunc_mu = True
        def iade_et(self):
            self.odunc_mu = False

        def __str__(self):
            durum = "[Ödünçte]" if self.odunc_mu else "[Rafta]"
            return f"{self.baslik} - {self.yazar} ({self.yil}) {durum}"

# Test:
#iki tane nesne oluşturuyoruz
kitap1 = Kitap("Suç ve Ceza", "Fyodor Dostoyeveski", 1866)
kitap2 = Kitap("Küçük Prens", "Antoine de Saint-Exupéry", 1943)

#bir tanesini ödünç veriyoruz
kitap2.odunc_ver()

#ikisini ekrana yazdırıp durumlarına bakıyoruz
print(kitap1)
print(kitap2)
````

### Soru 3: Hata Yönetimi

Özel hata sınıfları tanımla ve işlemleri güvenli hale getir. Soru 1 ve 2'deki metotları buna göre güncelle.

- Kendi hata sınıflarını yaz: `KitapBulunamadi` ve `KitapMusaitDegil` (`Exception` sınıfından türesin).
- `odunc_ver()` kitap zaten ödünçteyse, `iade_et()` kitap zaten raftaysa `KitapMusaitDegil` fırlatsın (`raise`).
- `kitap_bul()` kitabı bulamazsa `KitapBulunamadi` fırlatsın.
- Menüde bu hataları `try / except` ile yakala ve anlaşılır mesaj göster. Program çökmeksizin devam etsin.
- Yıl sayı olarak girilmezse (`ValueError`) uyarı ver.

````python
# --- Soru 3 Kod Alanı ---
class KitapBulunamadi(Exception):
    pass

class KitapMusaitDegil(Exception):
    pass

class Kitap:
    def __init__(self, baslik, yazar, yil):
        self.baslik = baslik
        self.yazar = yazar
        self.yil = yil
        self.odunc_mu = False

    def odunc_ver(self):
        if self.odunc_mu:
            raise KitapMusaitDegil(f"'{self.baslik}' zaten ödünç verilmiş!")
        self.odunc_mu = True

    def iade_et(self):
        if not self.odunc_mu:
            raise KitapMusaitDegil(f"'{self.baslik}' zaten rafta duruyor!")
        self.odunc_mu = False

    def __str__(self):
        durum = "[Ödünçte]" if self.odunc_mu else "[Rafta]"
        return f"{self.baslik} - {self.yazar} ({self.yil}) {durum}"



    class Kutuphane:
        def __init__(self):
            self.kitaplar = []

        def kitap_ekle(self, kitap):
            self.kitaplar.append(kitap)
        print(f"'{kitap.baslik}' kütüphaneye eklendi.")

    def kitap_bul(self, baslik):
        for kitap in self.kitaplar:
            if kitap.baslik.lower() == baslik.lower():
                return kitap
        # Bulunamazsa özel hata gönderiyoruz
        raise KitapBulunamadi(f"'{baslik}' isimli kitap bulunamadı!")

    def ara(self, kelime):
        sonuclar = [k for k in self.kitaplar if kelime.lower() in k.baslik.lower()]
        return sonuclar

    def odunc_ver(self, baslik):
        kitap = self.kitap_bul(baslik)  # Bulamazsa KitapBulunamadi gönderir
        kitap.odunc_ver()               # Ödünçteyse KitapMusaitDegil gönderir
        print(f"'{kitap.baslik}' başarıyla ödünç verildi.")

    def iade_al(self, baslik):
        kitap = self.kitap_bul(baslik)  # Bulamazsa KitapBulunamadi gönderir
        kitap.iade_et()                 # Raftaysa KitapMusaitDegil gönderir
        print(f"'{kitap.baslik}' başarıyla iade alındı.")



    kutuphane = Kutuphane()

while True:
    print("\n--- KÜTÜPHANE TAKİP SİSTEMİ ---")
    print("1. Kitap Ekle")
    print("2. Kitapları Listele")
    print("3. Kitap Ara")
    print("4. Kitap Ödünç Al")
    print("5. Kitap İade Et")
    print("6. Çıkış")

    secim = input("Seçiminiz (1-6): ").strip()

    if secim == "1":
        try:
            baslik = input("Kitap Başlığı: ").strip()
            yazar = input("Yazar: ").strip()
            yil_input = input("Yayın Yılı: ").strip()

        # Yıl sayı olarak girilmezse ValueError hatası gönderir
            yil = int(yil_input)
            
            yeni_kitap = Kitap(baslik, yazar, yil)
            kutuphane.kitap_ekle(yeni_kitap)

        except ValueError:
            print("Hata: Yayın yılı sadece sayılardan oluşmalıdır!")


    elif secim == "2":
        if not kutuphane.kitaplar:
            print("Kütüphanede henüz kitap yok.")
        else:
            print("\n--- Kütüphanedeki Kitaplar ---")
            for k in kutuphane.kitaplar:
                print(k)

    elif secim == "3":
        kelime = input("Aramak istediğiniz kelime: ").strip()
        bulunanlar = kutuphane.ara(kelime)
        if bulunanlar:
            print(f"\n'{kelime}' içeren kitaplar:")
            for k in bulunanlar:
                print(k)
        else:
            print("Hiçbir kitap bulunamadı.")

    elif secim == "4":
        try:
            baslik = input("Ödünç almak istediğiniz kitabın başlığı: ").strip()
            kutuphane.odunc_ver(baslik)
        except (KitapBulunamadi, KitapMusaitDegil) as e:
            print(f"Hata: {e}")

    elif secim == "5":
        try:
            baslik = input("İade etmek istediğiniz kitabın başlığı: ").strip()
            kutuphane.iade_al(baslik)
        except (KitapBulunamadi, KitapMusaitDegil) as e:
            print(f"Hata: {e}")

    elif secim == "6":
        print("Sistemden çıkılıyor... İyi günler!")
        break
        
    else:
        print("Geçersiz bir değer girdiniz, lütfen 1-6 arasında bir sayı girin.")

````

### Soru 4: Dosya İşlemleri (kitaplar.txt)

Kitapları kalıcı olarak dosyada sakla:

- Her kitap `kitaplar.txt` dosyasında bir satır olur. Alanlar `;` ile ayrılır, aralarında boşluk yoktur; son alan kitap ödünçteyse `1`, raftaysa `0` olur:
  ```text
  Suç ve Ceza;Fyodor Dostoyevski;1866;0
  Simyacı;Paulo Coelho;1988;1
  Kürk Mantolu Madonna;Sabahattin Ali;1943;0
  ```
- `Kutuphane` sınıfına kitapları dosyaya yazan `kaydet()` ve dosyadan okuyan `yukle()` metotlarını ekle.
- Program açılınca dosyadaki kitapları yükler. Dosya yoksa (`FileNotFoundError`) boş başlar.
- Menüdeki çıkış seçeneği `'Kaydet ve çık'` olsun ve kitapları dosyaya yazsın (`encoding='utf-8'`).
- Teslim etmeden önce dosyada en az 3 kitap olsun.

````python
# --- Soru 4 Kod Alanı ---
# Dosya okuma ve yazma metotlarını buraya ekleyiniz.
import os

# 1. Özel Hata Sınıfları
class KitapBulunamadi(Exception):
    pass

class KitapMusaitDegil(Exception):
    pass

# 2. Kitap Sınıfı
class Kitap:
    def __init__(self, baslik, yazar, yil, odunc_mu=False):
        self.baslik = baslik
        self.yazar = yazar
        self.yil = yil
        self.odunc_mu = odunc_mu

    def odunc_ver(self):
        if self.odunc_mu:
            raise KitapMusaitDegil(f"'{self.baslik}' zaten ödünç verilmiş!")
        self.odunc_mu = True

    def iade_et(self):
        if not self.odunc_mu:
            raise KitapMusaitDegil(f"'{self.baslik}' zaten rafta duruyor!")
        self.odunc_mu = False

    def __str__(self):
        durum = "[Ödünçte]" if self.odunc_mu else "[Rafta]"
        return f"{self.baslik} - {self.yazar} ({self.yil}) {durum}"

    # 3. Kutuphane Sınıfı (Dosya Kaydet/Yükle Özellikli)
class Kutuphane:
    def __init__(self, dosya_adi="kitaplar.txt"):
        self.kitaplar = []
        self.dosya_adi = dosya_adi

    def kitap_ekle(self, kitap):
        self.kitaplar.append(kitap)
        print(f"'{kitap.baslik}' kütüphaneye eklendi.")

    def kitap_bul(self, baslik):
        for kitap in self.kitaplar:
            if kitap.baslik.lower() == baslik.lower():
                return kitap
        raise KitapBulunamadi(f"'{baslik}' isimli kitap bulunamadı!")

    def ara(self, kelime):
        return [k for k in self.kitaplar if kelime.lower() in k.baslik.lower()]

    def odunc_ver(self, baslik):
        kitap = self.kitap_bul(baslik)
        kitap.odunc_ver()
        print(f"'{kitap.baslik}' başarıyla ödünç verildi.")

    def iade_al(self, baslik):
        kitap = self.kitap_bul(baslik)
        kitap.iade_et()
        print(f"'{kitap.baslik}' başarıyla iade alındı.")

    # Dosyaya Kaydetme Metodu
    def kaydet(self):
        with open(self.dosya_adi, "w", encoding="utf-8") as f:
            for k in self.kitaplar:
                odunc_durumu = "1" if k.odunc_mu else "0"
                satir = f"{k.baslik};{k.yazar};{k.yil};{odunc_durumu}\n"
                f.write(satir)
        print("Tüm veriler 'kitaplar.txt' dosyasına kaydedildi.")

    # Dosyadan Yükleme Metodu
    def yukle(self):
        try:
            with open(self.dosya_adi, "r", encoding="utf-8") as f:
                self.kitaplar = []
                for satir in f:
                    satir = satir.strip()
                    if satir:
                        baslik, yazar, yil, odunc_str = satir.split(";")
                        odunc_mu = True if odunc_str == "1" else False
                        kitap = Kitap(baslik, yazar, int(yil), odunc_mu)
                        self.kitaplar.append(kitap)
            print("Kitaplar dosyadan başarıyla yüklendi.")
        except FileNotFoundError:
            print("Kayıtlı dosya bulunamadı, yeni ve boş bir kütüphane oluşturuluyor.")

    # 4. Program Çalıştırma ve Menü
kutuphane = Kutuphane()
kutuphane.yukle()  

while True:
    print("\n--- KÜTÜPHANE TAKİP SİSTEMİ ---")
    print("1. Kitap Ekle")
    print("2. Kitapları Listele")
    print("3. Kitap Ara")
    print("4. Kitap Ödünç Al")
    print("5. Kitap İade Et")
    print("6. Kaydet ve Çık")
    
    secim = input("Seçiminiz (1-6): ").strip()

    if secim == "1":
        try:
            baslik = input("Kitap Başlığı: ").strip()
            yazar = input("Yazar: ").strip()
            yil_input = input("Yayın Yılı: ").strip()
            
            yil = int(yil_input)
            yeni_kitap = Kitap(baslik, yazar, yil)
            kutuphane.kitap_ekle(yeni_kitap)
        except ValueError:
            print("Hata: Yayın yılı sadece sayılardan oluşmalıdır!")

    elif secim == "2":
        if not kutuphane.kitaplar:
            print("Kütüphanede henüz kitap yok.")
        else:
            print("\n--- Kütüphanedeki Kitaplar ---")
            for k in kutuphane.kitaplar:
                print(k)

    elif secim == "3":
        kelime = input("Aramak istediğiniz kelime: ").strip()
        bulunanlar = kutuphane.ara(kelime)
        if bulunanlar:
            print(f"\n'{kelime}' içeren kitaplar:")
            for k in bulunanlar:
                print(k)
        else:
            print("Hiçbir kitap bulunamadı.")

    elif secim == "4":
        try:
            baslik = input("Ödünç almak istediğiniz kitabın başlığı: ").strip()
            kutuphane.odunc_ver(baslik)
        except (KitapBulunamadi, KitapMusaitDegil) as e:
            print(f"Hata: {e}")

    elif secim == "5":
        try:
            baslik = input("İade etmek istediğiniz kitabın başlığı: ").strip()
            kutuphane.iade_al(baslik)
        except (KitapBulunamadi, KitapMusaitDegil) as e:
            print(f"Hata: {e}")

    elif secim == "6":
        kutuphane.kaydet()  # Çıkarken dosyaya yazar
        print("Sistemden çıkılıyor... İyi günler!")
        break

    else:
        print("Geçersiz bir seçim yaptınız, lütfen 1-6 arasında bir sayı girin.")


````

### Soru 5: Iterator, Generator ve Kendi Modülün

Projenin son halini toparla:

- `Kutuphane` sınıfına `__iter__` metodunu ekle; böylece `for kitap in kutuphane:` yazılabilsin. Listelemeyi bu şekilde yap.
- Raftaki kitapları tek tek `yield` eden bir `musait_kitaplar()` generator metodu yaz.
- Menüye raftaki kitapları listeleyen yeni bir seçenek (`6) Raftakiler`) ekle.
- **Kendi modülün:** Hata sınıflarının, `Kitap` ve `Kutuphane` sınıflarının son halini aşağıdaki `%%writefile kutuphane.py` hücresine yaz. Hücre çalışınca kod `kutuphane.py` dosyasına kaydedilir. Menü hücresi sınıfları `from kutuphane import ...` ile alsın.

**Menü Seçenekleri:**
1) Kitap ekle | 2) Listele | 3) Ödünç ver | 4) İade al | 5) Ara | 6) Raftakiler | 0) Kaydet ve çık

```text
Seçimin: 6
Kürk Mantolu Madonna - Sabahattin Ali (1943) [Rafta]
Seçimin: 0
Kaydedildi. Görüşmek üzere!
```

````python
%%writefile kutuphane.py
# --- Soru 5: kutuphane.py ---
# Hata sınıflarının, Kitap ve Kutuphane sınıflarının son halini buraya yazınız.
# Bu hücreyi çalıştırınca kod kutuphane.py dosyasına kaydedilir.

# 1. Özel Hata Sınıfları
class KitapBulunamadi(Exception):
    pass

class KitapMusaitDegil(Exception):
    pass

# 2. Kitap Sınıfı
class Kitap:
    def __init__(self, baslik, yazar, yil, odunc_mu=False):
        self.baslik = baslik
        self.yazar = yazar
        self.yil = yil
        self.odunc_mu = odunc_mu

    def odunc_ver(self):
        if self.odunc_mu:
            raise KitapMusaitDegil(f"'{self.baslik}' zaten ödünç verilmiş.")
        self.odunc_mu = True

    def iade_et(self):
        if not self.odunc_mu:
            raise KitapMusaitDegil(f"'{self.baslik}' zaten rafta duruyor.")
        self.odunc_mu = False

    def __str__(self):
        durum = "[Ödünçte]" if self.odunc_mu else "[Rafta]"
        return f"{self.baslik} - {self.yazar} ({self.yil}) {durum}"


        # 3. Kutuphane Sınıfı
class Kutuphane:
    def __init__(self, dosya_adi="kitaplar.txt"):
        self.kitaplar = []
        self.dosya_adi = dosya_adi


    def __iter__(self):
        return iter(self.kitaplar)

    def musait_kitaplar(self):
        for kitap in self.kitaplar:
            if not kitap.odunc_mu:
                yield kitap

        def kitap_ekle(self, kitap):
        self.kitaplar.append(kitap)
        print(f"'{kitap.baslik}' eklendi.")

        def kitap_bul(self, baslik):
        for kitap in self.kitaplar:
            if kitap.baslik.lower() == baslik.lower():
                return kitap
        raise KitapBulunamadi(f"'{baslik}' bulunamadı.")

        def ara(self, kelime):
        return [k for k in self.kitaplar if kelime.lower() in k.baslik.lower()]

        def odunc_ver(self, baslik):
        kitap = self.kitap_bul(baslik)
        kitap.odunc_ver()
        print("Ödünç verildi.")

        def iade_al(self, baslik):
        kitap = self.kitap_bul(baslik)
        kitap.iade_et()
        print("İade alındı.")

        def kaydet(self):
        with open(self.dosya_adi, "w", encoding="utf-8") as f:
            for k in self.kitaplar:
                odunc_durumu = "1" if k.odunc_mu else "0"
                f.write(f"{k.baslik};{k.yazar};{k.yil};{odunc_durumu}\n")
        print("Kaydedildi. Görüşmek üzere!")

        def yukle(self):
        try:
            with open(self.dosya_adi, "r", encoding="utf-8") as f:
                self.kitaplar = []
                for satir in f:
                    satir = satir.strip()
                    if satir:
                        baslik, yazar, yil, odunc_str = satir.split(";")
                        odunc_mu = True if odunc_str == "1" else False
                        self.kitaplar.append(Kitap(baslik, yazar, int(yil), odunc_mu))
            print(f"{len(self.kitaplar)} kitap yüklendi.")
        except FileNotFoundError:
            print("Kayıtlı dosya bulunamadı, boş kütüphane oluşturuldu.")

````

````python
# --- Soru 5: Menü ---
# kutuphane.py değiştiğinde otomatik yeniden yüklensin:
%load_ext autoreload
%autoreload 2

from kutuphane import Kutuphane, Kitap, KitapBulunamadi, KitapMusaitDegil

# Menü kodunuzu buraya yazınız.
kutuphane = Kutuphane()
kutuphane.yukle()

while True:
    print("\n1. Kitap ekle | 2) Listele | 3) Ödünç ver | 4) İade al | 5) Ara | 6) Raftakiler | 0) Kaydet ve çık")
    
    secim = input("Seçimin: ").strip()

    if secim == "1":
        try:
            baslik = input("Kitap başlığı: ").strip()
            yazar = input("Yazar: ").strip()
            yil = int(input("Yayın yılı: ").strip())
            kutuphane.kitap_ekle(Kitap(baslik, yazar, yil))
        except ValueError:
            print("Hata: Yayın yılı sayı olmalıdır!")

    elif secim == "2":
        for kitap in kutuphane:
            print(kitap)

    elif secim == "3":
        try:
            baslik = input("Ödünç verilecek kitap: ").strip()
            kutuphane.odunc_ver(baslik)
        except (KitapBulunamadi, KitapMusaitDegil) as e:
            print(f"Hata: {e}")

    elif secim == "4":
        try:
            baslik = input("İade alınacak kitap: ").strip()
            kutuphane.iade_al(baslik)
        except (KitapBulunamadi, KitapMusaitDegil) as e:
            print(f"Hata: {e}")

    elif secim == "5":
        kelime = input("Arama kelimesi: ").strip()
        sonuclar = kutuphane.ara(kelime)
        for k in sonuclar:
            print(k)

    elif secim == "6":
        print("\n--- Raftaki (Ödünç Verilmeyen) Kitaplar ---")
        for k in kutuphane.musait_kitaplar():
            print(k)

    elif secim == "0":
        kutuphane.kaydet()
        break

    else:
        print("Geçersiz seçim!")
````
