### Soru 1: Katılımcı Kartı

Kullanıcıdan adını, soyadını, bölümünü ve doğum tarihini (gün, ay ve yıl ayrı ayrı) `input()` ile al, aşağıdaki gibi bir kart yazdır.

- Doğum tarihini `(gün, ay, yıl)` şeklinde bir **tuple**'da tut.
- Yaşı `2026 - doğum yılı` olarak hesapla.
- Ad soyadın baş harflerini büyüt, tarihi iki haneli yazdır (`07.03.2006`).

> **İpucu:** `f"{gun:02d}"` sayıyı iki haneye tamamlar.

**Örnek Çıktı Biçimi:**
```text
========================================
PyStart Katılımcı Kartı
========================================
Ad Soyad    : Ayşe Yılmaz
Bölüm       : Bilgisayar Mühendisliği
Doğum       : 07.03.2006
Yaş         : 20
========================================
```

````python
ad=input("Adınız:")
soyad=input("Soyadınız:")
bolum=input("Bölümünüz:")
gun=int(input("Doğum gününüz:"))
ay=int(input("Doğum ayınız:"))
yil=int(input("Doğum yılınız:"))
dogumTarihi=(gun,ay,yil)
yas=2026-dogumTarihi[2]
tamAd=f"{ad} {soyad}".title()
bolumFormatli=bolum.title()
tarihStr=f"{gun:02d},{ay:02d},{yil}"
print("="*30)
print("PyStart Katılımcı Kartı")
print("="*30)
print(f"Ad Soyad :{tamAd}")
print(f":{bolumFormatli}")
print(f"Doğum:{tarihStr}")
print(f"Yaş:{yas}")
print("="*30)
````

### Soru 2: Cümle Analizi

Kullanıcıdan bir cümle al ve şunları yazdır:

1. Karakter sayısı ve boşluklar hariç karakter sayısı
2. Kelime sayısı, ilk kelime ve son kelime
3. Cümlenin tamamen büyük harfli hali ve tersten yazılışı
4. Cümlede kaç tane 'a' harfi olduğu (büyük-küçük fark etmeden)
5. Cümlede 'python' kelimesinin geçip geçmediği (`True` ya da `False`)

> **Gözlem:** İçinde 'i' harfi olan bir cümlenin büyük harfli haline dikkatle bak. Ne fark ettin? Gözlemini yorum satırı olarak yaz.

````python
cumle = input("Bir cümle gir :")
karakterSayisi = len(cumle)
bosluksuz = len(cumle.replace(" ", ""))
kelimeler = cumle.split()
kelimeSayisi = len(kelimeler)
ilkKelime = kelimeler[0]
sonKelime = kelimeler[-1]
buyukCumle = cumle.upper()
tersten = cumle[::-1]
aSayisi = cumle.lower().count("a")
pythonVarMi = "python" in cumle.lower()
print("Karakter Sayısı:", karakterSayisi)
print("Boşluklar Hariç Karakter Sayısı:", bosluksuz)
print("Kelime Sayısı:", kelimeSayisi)
print("İlk Kelime:", ilkKelime)
print("Son Kelime:", sonKelime)
print("Büyük Harfli:", buyukCumle)
print("Tersten Yazılışı:", tersten)
print("Cümledeki'a' Sayısı:", aSayisi)
print("Python kelimesi geçiyor mu?:", pythonVarMi)

# Gözlem Yorumunuz: Türkçe küçük 'i' harfi büyük 'I' yerine İngilizce de büyük 'I' harfine dönüşür. Bu da Türkçe karakter uyumsuzluk yaratır.
````

### Soru 3: Not Listesi

Kullanıcıdan üç notu tek satırda, aralarında boşlukla al (`70 85 90`). Notları sayıya çevirip bir listeye koy, ardından:

1. Kullanıcıdan bir not daha alıp listeye ekle.
2. Listeyi küçükten büyüğe sırala ve yazdır.
3. En düşük notu, en yüksek notu ve ortalamayı yazdır.
4. En düşük notu listeden çıkar ve yeni ortalamayı yazdır.

> **İpucu:** `split()` metni boşluklardan böler ama parçalar hâlâ metindir.

**Örnek Çıktı:**
```text
3 notu aralarında boşluk bırakarak yaz: 70 85 90
Bir not daha ekle: 60
Notlar (sıralı): [60, 70, 85, 90]
En düşük: 60 | En yüksek: 90
Ortalama: 76.25
En düşük not çıkarıldı: [70, 85, 90]
Yeni ortalama: 81.67
```

````python
notGiris = input("3 notu aralarında boşluk bırakarak yaz :")
parcalar = notGiris.split()
notlar = [int(parcalar[0]),int(parcalar[1]),int(parcalar[2])]
yeniNot = int(input("Bir not daha ekle :"))
notlar.append(yeniNot)
notlar.sort()
print("Notlar (sıralı):", notlar)
print("En düşük:", min(notlar), "|En yüksek:", max(notlar))
print(f"Ortalama:{sum(notlar) / len(notlar)}")
notlar.remove(min(notlar))
print("En düşük not:", notlar)
print(f"Yeni ortalama:{sum(notlar) / len(notlar)}")

````

### Soru 5: Ortak Hobiler

İki kişinin hobilerini virgülle ayrılmış olarak al (`futbol,satranç,müzik`) ve her birini bir kümeye (`set`) çevir. Şunları yazdır:

1. Ortak hobiler, yalnızca birinci kişide olanlar, yalnızca ikinci kişide olanlar ve tüm hobiler
2. Şu soruların cevabı (`True` ya da `False` olarak):
   - Ortak hobiniz var mı?
   - Hobi sayılarınız eşit mi?
   - İkiniz de en az 3 hobi yazdınız mı?
   - 'satranç' ikinizden birinin hobisi mi?

> **İpucu:** Kümeleri yazdırırken sıra her çalıştırmada değişebilir; `sorted()` ile sabitleyebilirsin.

````python
hobi1 = input("1.Kişinin hobilerini giri:")
hobi2 = input("2. Kişinin hobilerini gir:")
kume1 = set(hobi1.split(","))
kume2 = set(hobi2.split(","))
ortak = kume1 & kume2
yalniz1 = kume1 - kume2
yalniz2 = kume2 - kume1
tumHobiler = kume1 | kume2
print("Ortak hobiler:", sorted(ortak))
print("Yalnızca ikinci kişide olanlar:", sorted(yalniz1))
print("Yalnızca ikinci kişide olanlar :", sorted(yalniz2))
print("Tüm hobiler :", sorted(tumHobiler))
print("Ortak hobiniz var mı?:", len(kume1 & kume2) > 0)
print("Hobi sayılarınız eşit mi?:", len(kume1) == len(kume2))
print("İkiniz de en az 3 hobi yazdınız mı?:", len(kume1) >= 3 and len(kume2) >= 3)
print("'satranç' ikinizden birini hobisi mi?:", "satranç" in tumHobiler)

````
