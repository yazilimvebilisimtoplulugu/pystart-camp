### Soru 3: Not Listesi

Kullanıcıdan üç notu tek satırda, aralarında boşlukla al (`70 85 90`). Notları sayıya çevirip bir listeye koy, ardından:

1. Kullanıcıdan bir not daha alıp listeye ekle.
2. Listeyi küçükten büyüğe sırala ve yazdır.
3. En düşük notu, en yüksek notu ve ortalamayı yazdır.
4. En düşük notu listeden çıkar ve yeni ortalamayı yazdır.

> **İpucu:** `split()` metni boşluklardan böler ama parçalar hâlâ metindir.

**Örnek Çıktı:**
```text
3 notu aralarında boşluk bırakarak yaz: 70 85 90
Bir not daha ekle: 60
Notlar (sıralı): [60, 70, 85, 90]
En düşük: 60 | En yüksek: 90
Ortalama: 76.25
En düşük not çıkarıldı: [70, 85, 90]
Yeni ortalama: 81.67
```

````python
# --- Soru 3 Kod Alanı ---
# Kodunuzu buraya yazınız.
notlar=input("3 notu tek satırda aralarında boşluk bırakarak giriniz:")
print("3 notu aralarında boşluk bırakarak yaz:",notlar)

notlar_listesi=notlar.split()
not1=int(notlar_listesi[0])
not2=int(notlar_listesi[1])
not3=int(notlar_listesi[2])
notlar_listesi=[not1,not2,not3]

not4=int(input("Bir not daha ekle:"))
print("Bir not daha ekle:",not4)
notlar_listesi.append(not4)


notlar_listesi.sort()
print("Sıralı notlar:",notlar_listesi)

en_dusuk=min(notlar_listesi)
en_yuksek=max(notlar_listesi)
ortalama=sum(notlar_listesi)/len(notlar_listesi)

print("En düşük not:",en_dusuk, "| En yüksek not:",en_yuksek)
print("Ortalama:",ortalama)

notlar_listesi.remove(en_dusuk)

yeni_ort= sum(notlar_listesi)/len(notlar_listesi)
print("En düşük not çıkarıldı:",notlar_listesi)
print("Yeni ortalama:",yeni_ort)


````

### Soru 4: Telefon Rehberi

Şu sözlükle başla:
```python
rehber = {'Ali': '0555 111 22 33', 'Ayşe': '0532 444 55 66'}
```

1. Kullanıcıdan bir isim ve numara alıp rehbere ekle.
2. Ali'nin numarasını `'0555 999 88 77'` olarak güncelle.
3. Ayşe'yi rehberden sil ve silinen numarayı yazdır.
4. Kullanıcıdan bir isim al ve numarasını yazdır. İsim rehberde yoksa program hata vermesin, `'Kayıt bulunamadı'` yazsın.
5. Rehberdeki kişi sayısını ve isimleri alfabetik sırayla yazdır.

> **İpucu:** Sözlüklerin `get()` metoduna ikinci bir varsayılan değer verebilirsin.

````python
rehber = {'Ali': '0555 111 22 33', 'Ayşe': '0532 444 55 66'}

# --- Soru 4 Kod Alanı ---
# Kodunuzu buraya yazınız.
isim=input("Rehbere eklenecek ismi giriniz: ")
telefon=input("Telefon numarsını giriniz: ")

rehber[isim]=telefon

rehber["Ali"]="0555 999 88 77"

silinecek_numara=rehber.pop("Ayşe")
print("Ayşe'nin silinen numarası:",silinecek_numara)

aranan_isim=input("Aramak istediğiniz ismi giriniz:",)
numara=rehber.get(aranan_isim,'Kayıt Bulunamadı')
print("Aranan ismin numarası:",numara)

kisi_sayisi=len(rehber)
alfabetik_isimler=sorted(rehber.keys())
print("Rehberdeki kişi sayısı:",kisi_sayisi)
print("Alfabetik isim listesi:",alfabetik_isimler)
````

### Soru 5: Ortak Hobiler

İki kişinin hobilerini virgülle ayrılmış olarak al (`futbol,satranç,müzik`) ve her birini bir kümeye (`set`) çevir. Şunları yazdır:

1. Ortak hobiler, yalnızca birinci kişide olanlar, yalnızca ikinci kişide olanlar ve tüm hobiler
2. Şu soruların cevabı (`True` ya da `False` olarak):
   - Ortak hobiniz var mı?
   - Hobi sayılarınız eşit mi?
   - İkiniz de en az 3 hobi yazdınız mı?
   - 'satranç' ikinizden birinin hobisi mi?

> **İpucu:** Kümeleri yazdırırken sıra her çalıştırmada değişebilir; `sorted()` ile sabitleyebilirsin.

````python
# --- Soru 5 Kod Alanı ---
# Kodunuzu buraya yazınız.

hobi1=input("1.kişinin hobilerini virgülle ayırarak yazın: ")
hobi2=input("2.kişinin hobilerini virgülle ayırarak yazın: ")

kume1= set(hobi1.lower().replace(" ","").split(","))
kume2= set(hobi2.lower().replace(" ","").split(","))

ortak_hobiler= kume1 & kume2
sadece_kisi1_hobileri= kume1 - kume2
sadece_kisi2_hobileri= kume2 - kume1
tum_hobiler= kume1 | kume2

print("ortak hobiler:",sorted(ortak_hobiler))
print("Sadece 1.kişinin hobileri:",sorted(sadece_kisi1_hobileri))
print("Sadece 2.kişinin hobileri:",sorted(sadece_kisi2_hobileri))
print("Tüm hobiler:",sorted(tum_hobiler))

print("Ortak hobiniz var mı?:",len(ortak_hobiler)>0)
print("Hobi sayılarınız eşit mi?:",len(kume1)==len(kume2))
print(len(kume1)>=3 and len(kume2)>=3)
print("'satranç' ikinizden birinin hobisi mi?:","satranç" in tum_hobiler)

````
