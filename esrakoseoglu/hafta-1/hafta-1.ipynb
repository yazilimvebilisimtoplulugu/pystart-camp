```python
#KATILIMCI KARTI

ad = input("Adınız: ")
soyad = input("Soyadınız: ")
bolum = input("Bölümünüz: ")

dogum_girdisi = input("Doğum tarihiniz (GG.AA.YYYY): ")

gun, ay, yil = map(int, dogum_girdisi.split("."))
dogum_tarihi = (gun, ay, yil)

yas = 2026 - dogum_tarihi[2]

tam_ad = ad.title() + " " + soyad.title()
tarih = f"{gun:02d}.{ay:02d}.{yil}"

print("=" * 40)
print("PyStart Katılımcı Kartı")
print("=" * 40)
print(f"Ad Soyad   : {tam_ad}")
print(f"Bölüm      : {bolum}")
print(f"Doğum      : {tarih}")
print(f"Yaş        : {yas}")
print("=" * 40)
```

```python
#CÜMLE ANALİZİ

cumle = input("Bir cümle yazınız: ")

karakter_sayisi = len(cumle)
bosluksuz_karakter = len(cumle.replace(" ", ""))

print("Karakter sayısı:", karakter_sayisi)
print("Boşluklar hariç karakter sayısı:", bosluksuz_karakter)

kelimeler = cumle.split()

print("Kelime sayısı:", len(kelimeler))
print("İlk kelime:", kelimeler[0])
print("Son kelime:", kelimeler[-1])

print("Büyük harfli hali:", cumle.upper())
print("Tersten yazılışı:", cumle[::-1])

print("'a' harfi sayısı:", cumle.lower().count("a"))

print("Python kelimesi geçiyor mu:", "python" in cumle.lower())

# Gözlem Yorumunuz:
# upper() metodu İngilizce 'i' harfini 'I' yapar.
```

```python
#NOT LİSTESİ

notlar = input("3 notu aralarında boşluk bırakarak yaz: ").split()

notlar = [int(not_degeri) for not_degeri in notlar]

ek_not = int(input("Bir not daha ekle: "))
notlar.append(ek_not)

notlar.sort()
print("Notlar (sıralı):", notlar)

en_dusuk = min(notlar)
en_yuksek = max(notlar)
ortalama = sum(notlar) / len(notlar)

print("En düşük:", en_dusuk, "| En yüksek:", en_yuksek)
print(f"Ortalama: {ortalama:.2f}")

notlar.remove(en_dusuk)

yeni_ortalama = sum(notlar) / len(notlar)

print("En düşük not çıkarıldı:", notlar)
print(f"Yeni ortalama: {yeni_ortalama:.2f}")
```

```python
#TELEFON REHBERİ

rehber = {
    'Ali': '0555 111 22 33',
    'Ayşe': '0532 444 55 66'
}

yeni_isim = input("Eklenecek kişinin adı: ")
yeni_numara = input("Telefon numarası: ")
rehber[yeni_isim] = yeni_numara

rehber['Ali'] = '0555 999 88 77'

silinen_numara = rehber.pop('Ayşe')
print("Silinen numara:", silinen_numara)

aranan_isim = input("Numarasını öğrenmek istediğiniz kişi: ")
print(rehber.get(aranan_isim, 'Kayıt bulunamadı'))

print("Rehberdeki kişi sayısı:", len(rehber))
print("İsimler:", sorted(rehber.keys()))
```

```python
#ORTAK HOBİLER

hobiler1 = set(input("Birinci kişinin hobileri (virgülle ayır): ").lower().replace(" ", "").split(","))
hobiler2 = set(input("İkinci kişinin hobileri (virgülle ayır): ").lower().replace(" ", "").split(","))

ortak_hobiler = hobiler1 & hobiler2
yalniz_birinci = hobiler1 - hobiler2
yalniz_ikinci = hobiler2 - hobiler1
tum_hobiler = hobiler1 | hobiler2

print("Ortak hobiler:", sorted(ortak_hobiler))
print("Yalnızca birinci kişide olanlar:", sorted(yalniz_birinci))
print("Yalnızca ikinci kişide olanlar:", sorted(yalniz_ikinci))
print("Tüm hobiler:", sorted(tum_hobiler))

print("Ortak hobiniz var mı?:", bool(ortak_hobiler))
print("Hobi sayılarınız eşit mi?:", len(hobiler1) == len(hobiler2))
print("İkiniz de en az 3 hobi yazdınız mı?:", len(hobiler1) >= 3 and len(hobiler2) >= 3)
print("'satranç' ikinizden birinin hobisi mi?:", "satranç" in hobiler1 or "satranç" in hobiler2)
```
