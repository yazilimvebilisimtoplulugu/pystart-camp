### Soru 1: Katılımcı Kartı

Kullanıcıdan adını, soyadını, bölümünü ve doğum tarihini (gün, ay ve yıl ayrı ayrı) `input()` ile al, aşağıdaki gibi bir kart yazdır.

- Doğum tarihini `(gün, ay, yıl)` şeklinde bir **tuple**'da tut.
- Yaşı `2026 - doğum yılı` olarak hesapla.
- Ad soyadın baş harflerini büyüt, tarihi iki haneli yazdır (`07.03.2006`).

> **İpucu:** `f"{gun:02d}"` sayıyı iki haneye tamamlar.

**Örnek Çıktı Biçimi:**
```text
========================================
PyStart Katılımcı Kartı
========================================
Ad Soyad    : Ayşe Yılmaz
Bölüm       : Bilgisayar Mühendisliği
Doğum       : 07.03.2006
Yaş         : 20
========================================
```

````python
# bilgi alma
ad = input("Ad: ")
soyad = input("Soyad: ")
bolum = input("Bölüm: ")

# doğum tarihini ayrı ayrı alma
gun = int(input("Doğum günü: "))
ay = int(input("Doğum ayı: "))
yıl = int(input("Doğum yılı: "))
dogum_tarihi = (gun,ay,yıl)
yas = 2026 -dogum_tarihi [2]

#ilk harfi büyük yapma
ad = ad.capitalize()
soyad = soyad.capitalize()

#en az 2 basamakla tarihi yazdırma
dogum_tarihi = f"{dogum_tarihi[0]:02d}.{dogum_tarihi[1]:02d}.{dogum_tarihi[2]}"

# kart şablon çıktısı
print("=======================")
print("PyStart Katılımcı Kartı")
print("=======================")
print(f"Ad Soyad: {ad} {soyad}")
print(f"Bölüm: {bolum}")
print(f"Doğum Tarihi: {dogum_tarihi}")
print(f"Yaş: {yas}")
print("=======================")

````

### Soru 3: Not Listesi

Kullanıcıdan üç notu tek satırda, aralarında boşlukla al (`70 85 90`). Notları sayıya çevirip bir listeye koy, ardından:

1. Kullanıcıdan bir not daha alıp listeye ekle.
2. Listeyi küçükten büyüğe sırala ve yazdır.
3. En düşük notu, en yüksek notu ve ortalamayı yazdır.
4. En düşük notu listeden çıkar ve yeni ortalamayı yazdır.

> **İpucu:** `split()` metni boşluklardan böler ama parçalar hâlâ metindir.

**Örnek Çıktı:**
```text
3 notu aralarında boşluk bırakarak yaz: 70 85 90

En düşük: 60 | En yüksek: 90
Ortalama: 76.25
En düşük not çıkarıldı: [70, 85, 90]
Yeni ortalama: 81.67
```

### Soru 4: Telefon Rehberi

Şu sözlükle başla:
```python
rehber = {'Ali': '0555 111 22 33', 'Ayşe': '0532 444 55 66'}
```

1. Kullanıcıdan bir isim ve numara alıp rehbere ekle.
2. Ali'nin numarasını `'0555 999 88 77'` olarak güncelle.
3. Ayşe'yi rehberden sil ve silinen numarayı yazdır.
4. Kullanıcıdan bir isim al ve numarasını yazdır. İsim rehberde yoksa program hata vermesin, `'Kayıt bulunamadı'` yazsın.
5. Rehberdeki kişi sayısını ve isimleri alfabetik sırayla yazdır.

> **İpucu:** Sözlüklerin `get()` metoduna ikinci bir varsayılan değer verebilirsin.

````python
rehber = {'Ali': '0555 111 22 33', 'Ayşe': '0532 444 55 66'}
isim = input("İsim giriniz:")
numara = input("Numara giriniz:")

rehber[isim]= numara
rehber["Ali"] = "0555 999 88 77"
sil_numara= rehber.pop("Ayşe")
print("Silinen numara:", sil_numara)
isim = input("İsim giriniz:")
#get fonksiyonuyla aradığımız ismi buluyoruz.
print("Numara:", rehber.get(isim, "Kayıt bulunamadı"))
print("Kişi sayısı:", len(rehber))
print("İsimler sırasıyla:", sorted(rehber))



````

### Soru 5: Ortak Hobiler

İki kişinin hobilerini virgülle ayrılmış olarak al (`futbol,satranç,müzik`) ve her birini bir kümeye (`set`) çevir. Şunları yazdır:

1. Ortak hobiler, yalnızca birinci kişide olanlar, yalnızca ikinci kişide olanlar ve tüm hobiler
2. Şu soruların cevabı (`True` ya da `False` olarak):
   - Ortak hobiniz var mı?
   - Hobi sayılarınız eşit mi?
   - İkiniz de en az 3 hobi yazdınız mı?
   - 'satranç' ikinizden birinin hobisi mi?

> **İpucu:** Kümeleri yazdırırken sıra her çalıştırmada değişebilir; `sorted()` ile sabitleyebilirsin.

````python
kisi1_hobi = input("Birinci kişinin hobilerini virgülle ayrılmış olarak giriniz:")
kisi2_hobi = input("İkinci kişinin hobilerini virgülle ayrılmış olarak giriniz:")
#splpit metotduyla liste şekline çevirdik.
kisi1=set(kisi1_hobi.split(","))
kisi2=set(kisi2_hobi.split(","))
#sorted sabitlemesi
print("Ortak hobiler:" ,  sorted(kisi1 & kisi2))
print("Yalnızca birinci kişide olanlar:", sorted(kisi1 - kisi2))
print("Yalnızca ikinci kişide olanlar:", sorted(kisi2 - kisi1))
print("Tüm hobiler:", sorted(kisi1 | kisi2)) # birleşim
print("Ortak hobiniz var mı?:", bool(kisi1 & kisi2))
print("Hobi sayılarınız eşit mi?:", len(kisi1) == len(kisi2))
print("İkiniz de en az 3 hobi yazdınız mı?:", len(kisi1) >= 3 and len(kisi2) >= 3)
print("satranç ikinizden birinin hobisi mi?:", "satranç" in kisi1 or "satranç" in kisi2)

````
