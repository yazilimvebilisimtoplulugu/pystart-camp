### Soru 3: Not Listesi

Kullanıcıdan üç notu tek satırda, aralarında boşlukla al (`70 85 90`). Notları sayıya çevirip bir listeye koy, ardından:

1. Kullanıcıdan bir not daha alıp listeye ekle.
2. Listeyi küçükten büyüğe sırala ve yazdır.
3. En düşük notu, en yüksek notu ve ortalamayı yazdır.
4. En düşük notu listeden çıkar ve yeni ortalamayı yazdır.

> **İpucu:** `split()` metni boşluklardan böler ama parçalar hâlâ metindir.

**Örnek Çıktı:**
```text
3 notu aralarında boşluk bırakarak yaz: 70 85 90
Bir not daha ekle: 60
Notlar (sıralı): [60, 70, 85, 90]
En düşük: 60 | En yüksek: 90
Ortalama: 76.25
En düşük not çıkarıldı: [70, 85, 90]
Yeni ortalama: 81.67
```

````python
# --- Soru 3 Kod Alanı ---
# Kodunuzu buraya yazınız.

#Kullanıcıdan 3 not al ve liste oluştur.
notlar = input("Üç not giriniz, lütfen aralarında boşluk bırakmayı unutmayınız: ")

notlar_list = list(map(int, notlar.split()))
print("Notlar listesi: ", notlar_list)
#Kullanıcıdan bir not daha alıp listeye ekleyiniz.
bir_not_daha=int(input("Bir not daha giriniz lütfen: "))
notlar_list.append(bir_not_daha)
print("Notlar listesi güncellendi: ",notlar_list)
#Listeyi küçükten büyüğe sırala ve yazdır.
sirali_notlar_listesi=sorted(notlar_list)
print("Sıralanmış notlar listesi: ",sirali_notlar_listesi)
#En düşük notu yazdır.
min_not=min(notlar_list)
print("En düşük not: ",min_not)
#En yüksek notu yazdır.
max_not=max(notlar_list)
print("En yüksek not: ",max_not)
#Ortalamayı yazdır.
avarage_notlar=sum(notlar_list)/len(notlar_list)
print("Not ortalaması: ",avarage_notlar)
#En düşük notu listeden çıkart ve yeni ortalamayı yazdır.

sirali_notlar_listesi.pop(0)


iyi_not_ortalamasi = sum(sirali_notlar_listesi) / len(sirali_notlar_listesi)
print("Minimum not olmadan ortalama: ", iyi_not_ortalamasi)
````

### Soru 4: Telefon Rehberi


1. Kullanıcıdan bir isim ve numara alıp rehbere ekle.
2. Ali'nin numarasını `'0555 999 88 77'` olarak güncelle.
3. Ayşe'yi rehberden sil ve silinen numarayı yazdır.
4. Kullanıcıdan bir isim al ve numarasını yazdır. İsim rehberde yoksa program hata vermesin, `'Kayıt bulunamadı'` yazsın.
5. Rehberdeki kişi sayısını ve isimleri alfabetik sırayla yazdır.

> **İpucu:** Sözlüklerin `get()` metoduna ikinci bir varsayılan değer verebilirsin.

````python
rehber = {'Ali': '0555 111 22 33', 'Ayşe': '0532 444 55 66'}

# --- Soru 4 Kod Alanı ---
# Kodunuzu buraya yazınız.

#Kullanıcıdan bir isim ve numara alıp rehbere ekle.
rehber={"Ali":"0555 111 22 33", "Ayşe":"0532 444 55 66"}
isim=input("İsim giriniz: ")
numara=input("Telefon numarası giriniz:")
rehber[isim]=numara
print("rehber:", rehber)

#Ali'nin numarasını "0555 999 88 77" olarak güncelle.
rehber["Ali"]="0555 999 88 77"
print("Güncel rehber: ",rehber)
#Ayşe'yi rehberden sil ve silinen numarayı yazdır.
silinen_kişi=rehber.pop("Ayşe")
print("Silinen numara: ",silinen_kişi)

#Kullanıcıdan bir isim al ve numarasını yazdır. İsim rehberde yoksa program hata vermesin, "Kayıt bulunamadı" yazsın.
kisi_ismi=input("Kimi arıyorsunuz: ")
kisi_no=input("Hangi telefon numarasını arıyorsunuz: ")
if kisi_ismi in rehber:
    if kisi_no==rehber[kisi_ismi]:
        print("Kayıt rehberde mevcut")
    else:
        print("Kayıt bulunamadı")
#Rehberdeki kişi sayısını ve isimlerini alfabetik sırayla yazdır.
kisi_adeti=len(rehber)
print("Kişi sayısı: ",kisi_adeti)
sirali_rehber=sorted(rehber)
print("Alfabetik düzende rehber: ")
for x in rehber:
    print(x)
````

### Soru 5: Ortak Hobiler

İki kişinin hobilerini virgülle ayrılmış olarak al (`futbol,satranç,müzik`) ve her birini bir kümeye (`set`) çevir. Şunları yazdır:

1. Ortak hobiler, yalnızca birinci kişide olanlar, yalnızca ikinci kişide olanlar ve tüm hobiler
2. Şu soruların cevabı (`True` ya da `False` olarak):
   - Ortak hobiniz var mı?
   - Hobi sayılarınız eşit mi?
   - İkiniz de en az 3 hobi yazdınız mı?
   - 'satranç' ikinizden birinin hobisi mi?

> **İpucu:** Kümeleri yazdırırken sıra her çalıştırmada değişebilir; `sorted()` ile sabitleyebilirsin.

````python
# --- Soru 5 Kod Alanı ---
# Kodunuzu buraya yazınız.

hobiler_a=input("Hobilerinizi virgülle ayrılmış bir biçimde giriniz: ")
hobiler_b=input("Hobilerinizi virgülle ayrılmış bir biçimde giriniz: ")

kume_a=set(hobiler_a.split(","))
kume_b=set(hobiler_b.split(","))

#Ortak hobiler
ortak= kume_a & kume_b
print("Ortak hobiler: ",ortak)

#Sadece A kişisinin hobileri
sadece_a= kume_a - kume_b
print("Sadece 1. kişiye ait hobiler: ", sadece_a)
 
 #Sadece B kişisinin hobileri
sadece_b= kume_b - kume_a
print("Sadece 2. kişiye ait hobiler: ", sadece_b)

#Tüm hobiler
hepsi=kume_a|kume_b
print("Tüm hobiler: ", hepsi)

#Soruların cevapları

#Ortak hobiniz var mı?
print("Ortak hobiniz var mı?:", "Evet" if len(ortak) > 0 else "Hayır")
print("Hobi sayılarınız eşit mi?", "Evet" if len(kume_b)==len(kume_a) else "Hayır")
print("İkiniz de en az üç hobi yazdınız mı?", "Evet" if len(kume_a)>=3 and len(kume_b)>=3 else "Hayır")
print(" satranç ikinizden birinin hobisi mi?", "Evet" if "satranç" in kume_a or "satranç" in kume_b else "Hayır" )
    










 
````
