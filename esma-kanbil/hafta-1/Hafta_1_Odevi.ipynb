### Soru 1: Katılımcı Kartı

Kullanıcıdan adını, soyadını, bölümünü ve doğum tarihini (gün, ay ve yıl ayrı ayrı) `input()` ile al, aşağıdaki gibi bir kart yazdır.

- Doğum tarihini `(gün, ay, yıl)` şeklinde bir **tuple**'da tut.
- Yaşı `2026 - doğum yılı` olarak hesapla.
- Ad soyadın baş harflerini büyüt, tarihi iki haneli yazdır (`07.03.2006`).

> **İpucu:** `f"{gun:02d}"` sayıyı iki haneye tamamlar.

**Örnek Çıktı Biçimi:**
```text
========================================
PyStart Katılımcı Kartı
========================================
Ad Soyad    : Ayşe Yılmaz
Bölüm       : Bilgisayar Mühendisliği
Doğum       : 07.03.2006
Yaş         : 20
========================================
```

````python
# 1. Kullanıcıdan adını, soyadını, bölümünü alma
ad = input("Adınızı girin: ")
soyad = input("Soyadınızı girin: ")
bolum = input("Bölümünüzü girin: ")

# 2. Kullanıcıdan doğum tarihini alma
gun = int(input("Doğum günü: "))
ay = int(input("Doğum ayı: "))
yil = int(input("Doğum yılı: "))

# 3. Doğum tarihini tuple'da tutma ve iki haneli yazma
dogum_tarihi_str = (gun, ay, yil)
dogum_tarihi = f"{gun:02d}.{ay:02d}.{yil}"

# 4. Yaş ve Ad soyad biçimlendirme
yas = 2026 - dogum_tarihi_str[2]  # ikinci indis yani 3. eleman bize yılı verir 2026 - yıl olur
ad_soyad = f"{ad} {soyad}".title()
bolum = f"{bolum}".title()

# 5. kart çıktısı 
print("======================")
print("PyStart Katılımcı Kartı")
print("======================")
print(f"Ad Soyad     : {ad_soyad}")
print(f"Bölüm        : {bolum}")
print(f"Doğum Tarihi : {dogum_tarihi}")
print(f"Yaş          : {yas}")
print("======================")

````

### Soru 2: Cümle Analizi

Kullanıcıdan bir cümle al ve şunları yazdır:

1. Karakter sayısı ve boşluklar hariç karakter sayısı
2. Kelime sayısı, ilk kelime ve son kelime
3. Cümlenin tamamen büyük harfli hali ve tersten yazılışı
4. Cümlede kaç tane 'a' harfi olduğu (büyük-küçük fark etmeden)
5. Cümlede 'python' kelimesinin geçip geçmediği (`True` ya da `False`)

> **Gözlem:** İçinde 'i' harfi olan bir cümlenin büyük harfli haline dikkatle bak. Ne fark ettin? Gözlemini yorum satırı olarak yaz.

````python
# Kullanıcıdan bir cümle alma
cumle = input("Bir cümle giriniz: ")

# 1. Karakter sayısı 
karakter_sayisi = len(cumle)
bosluksuz_karakter_sayisi = len(cumle.replace(""," "))

# 2. Kelime 
kelimeler = cumle.split()
kelime_sayisi = len(kelimeler)
ilk_kelime = kelimeler[0]
son_kelime = kelimeler[-1]

# 3. büyük harf ve tersten yazılış
cumle_buyuk = cumle.upper()
cumle_ters = cumle[::-1]

# 4. kaç a harfi var
a_harfi_sayisi = cumle.lower().count("a")

# 5. python kelimesi geçiyor mu 
python_TF = "python" in cumle.lower()

# çıktılar
print(f"Karakter Sayısı : {karakter_sayisi}")
print(f"Boşluksuz Karakter Sayısı : {bosluksuz_karakter_sayisi}")
print(f"Kelime Sayısı : {kelime_sayisi}")
print(f"İlk Kelime : {ilk_kelime}")
print(f"Son Kelime : {son_kelime}")
print(f"Büyük Harfli Cümle : {cumle_buyuk}")
print(f"Tersten Yazılışı : {cumle_ters}")
print(f"Cümledeki 'a' harfi sayısı : {a_harfi_sayisi}")
print(f"'python' geçiyor mu : {python_TF}")

# Gözlem Yorumunuz: 
# Küçük i harfi normal olarak var ama python ingilizce alfabede iyi çalıştığı için büyük i harfi I oluyor. 
# Ancak direkt büyük İ kullandıysak python zaten buna dokunmuyor o olduğu gibi kalıyor.


````

### Soru 3: Not Listesi

Kullanıcıdan üç notu tek satırda, aralarında boşlukla al (`70 85 90`). Notları sayıya çevirip bir listeye koy, ardından:

1. Kullanıcıdan bir not daha alıp listeye ekle.
2. Listeyi küçükten büyüğe sırala ve yazdır.
3. En düşük notu, en yüksek notu ve ortalamayı yazdır.
4. En düşük notu listeden çıkar ve yeni ortalamayı yazdır.

> **İpucu:** `split()` metni boşluklardan böler ama parçalar hâlâ metindir.

**Örnek Çıktı:**
```text
3 notu aralarında boşluk bırakarak yaz: 70 85 90
Bir not daha ekle: 60
Notlar (sıralı): [60, 70, 85, 90]
En düşük: 60 | En yüksek: 90
Ortalama: 76.25
En düşük not çıkarıldı: [70, 85, 90]
Yeni ortalama: 81.67
```

````python
# 1. Kullanıcıdan 3 not alma
notlar_str = input(" 3 notu aralarında boşluk bırakarak yaz: ").split()
notlar = [int(notlar_str[0]), int(notlar_str[1]), int(notlar_str[2])]

# 2. Bir not daha alıp listeye ekleme
ek_not = int(input("Bir not daha ekle: "))
notlar.append(ek_not)

# 3. Listeyi küçükten büyüğe sıralama
notlar.sort()
print(f"Notlar sıralı : {notlar}")

# 4. En düşük, en yüksek ve ortalama hesaplama
en_dusuk = min(notlar)   # sıralı olduğu için notlar[0] da olur
en_yuksek = max(notlar)  # sıralı olduğu için notlar[-1] de olur
ortalama = sum(notlar) / len(notlar)
print(f"En düşük : {en_dusuk}")
print(f"En yüksek : {en_yuksek}")
print(f"Ortalama : {ortalama} ")

# 5. En düşük notu listeden çıkartma ve yeni ortalama
notlar.pop(0)
yeni_ortalama = sum(notlar) / len(notlar)
print(f"En düşük not çıkarıldı : {notlar}")
print(f"Yeni ortalama : {yeni_ortalama}")
````

### Soru 4: Telefon Rehberi

Şu sözlükle başla:
```python
rehber = {'Ali': '0555 111 22 33', 'Ayşe': '0532 444 55 66'}
```

1. Kullanıcıdan bir isim ve numara alıp rehbere ekle.
2. Ali'nin numarasını `'0555 999 88 77'` olarak güncelle.
3. Ayşe'yi rehberden sil ve silinen numarayı yazdır.
4. Kullanıcıdan bir isim al ve numarasını yazdır. İsim rehberde yoksa program hata vermesin, `'Kayıt bulunamadı'` yazsın.
5. Rehberdeki kişi sayısını ve isimleri alfabetik sırayla yazdır.

> **İpucu:** Sözlüklerin `get()` metoduna ikinci bir varsayılan değer verebilirsin.

````python
rehber = {'Ali': '0555 111 22 33', 'Ayşe': '0532 444 55 66'}

# 1. Kullanıcıdan isim ve numara alma, rehbere ekleme
isim = input("Lütfen bir isim giriniz: ")
numara = input("Lütfen bir telefon numarası giriniz: ")
rehber[isim] = numara

# 2. Alinin no'yu güncelle 
rehber['Ali'] = '0555 999 88 77'

# 3. Ayşeyi sil  ve silinen numarayı yazdır
silinen_numara = rehber.pop('Ayşe')
print(f"Ayşe rehberden silindi. Silinen numara: {silinen_numara}")

# 4. Kullanıcıdan isim al no yazdır yoksa kayıt bulunamadı yazsın
aranan_isim = input("Aramak istediğiniz ismi giriniz: ")
print("Numara: ", rehber.get(aranan_isim, "Kayıt bulunamadı"))

# 5. Rehberdeki kişi sayısı ve alfabetik sıralama 
kisi_sayisi = len(rehber)
isimler_sirali = sorted(rehber.keys())

print(f"Rehberdeki toplam kişi sayısı: {kisi_sayisi}")
print(f"Kişiler (alfabetik): {isimler_sirali}")


````

### Soru 5: Ortak Hobiler

İki kişinin hobilerini virgülle ayrılmış olarak al (`futbol,satranç,müzik`) ve her birini bir kümeye (`set`) çevir. Şunları yazdır:

1. Ortak hobiler, yalnızca birinci kişide olanlar, yalnızca ikinci kişide olanlar ve tüm hobiler
2. Şu soruların cevabı (`True` ya da `False` olarak):
   - Ortak hobiniz var mı?
   - Hobi sayılarınız eşit mi?
   - İkiniz de en az 3 hobi yazdınız mı?
   - 'satranç' ikinizden birinin hobisi mi?

> **İpucu:** Kümeleri yazdırırken sıra her çalıştırmada değişebilir; `sorted()` ile sabitleyebilirsin.

````python
# 1. Kullancıdan hobileri alıp set yapma
hobi1 = input("1. kişinin hobileri: ").split(",")
hobi2 = input("2. kişinin hobileri: ").split(",")

kisi1 = set(hobi1)
kisi2 = set(hobi2)

# Ortak hobi, yalnızca 1, yalnızca 2 de olanlar
ortak_hobiler = kisi1 & kisi2
just_1 = kisi1 - kisi2
just_2 = kisi2 - kisi1
tum_hobiler = kisi1 | kisi2 

print(f"Ortak Hobiler         :  {sorted(ortak_hobiler)}")
print(f"Yalnızca 1 de olanlar :  {sorted(just_1)}")
print(f"Yalnızca 2 de olanlar :  {sorted(just_2)}")
print(f"Tüm hobiler           :  {sorted(tum_hobiler)}")

print(f"Ortak hobiniz var mı?                     :  {len(ortak_hobiler) > 0}" )
print(f"Hobi sayılarınız eşit mi?                 :  {len(kisi1) == len(kisi2)} ")
print(f"İkinizde en az 3 hobi yazdınız mı?        :  {len(kisi1) >= 3 and len(kisi2) >= 3}" )
print(f"'satranç' ikinizden birinin hobisi mi?    :  {'satranç' in tum_hobiler}" )
````
