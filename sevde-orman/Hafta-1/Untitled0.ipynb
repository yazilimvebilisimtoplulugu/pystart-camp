Soru 1: Katılımcı Kartı
Kullanıcıdan adını, soyadını, bölümünü ve doğum tarihini (gün, ay ve yıl ayrı ayrı) input() ile al, aşağıdaki gibi bir kart yazdır.

Doğum tarihini (gün, ay, yıl) şeklinde bir tuple'da tut.
Yaşı 2026 - doğum yılı olarak hesapla.
Ad soyadın baş harflerini büyüt, tarihi iki haneli yazdır (07.03.2006).
İpucu: f"{gun:02d}" sayıyı iki haneye tamamlar.

Örnek Çıktı Biçimi:

========================================
PyStart Katılımcı Kartı
========================================
Ad Soyad    : Ayşe Yılmaz
Bölüm       : Bilgisayar Mühendisliği
Doğum       : 07.03.2006
Yaş         : 20
========================================


```python

# Bilgileri kullanicidan al
ad = input("Isminiz: ")
soyad = input("Soyisminiz: ")
bolum = input("Bolumunuz: ")

gun = int(input("Dogdugunuz gun: "))
ay = int(input("Dogdugunuz ay: "))
yil = int(input("Dogdugunuz yil: "))


dogum = (gun, ay, yil)

# Yas hesapla
yas = 2026 - yil

# Bas harfleri buyut
ad_soyad = ad.capitalize() + " " + soyad.capitalize()

# Kartı ekrana yazdir
print("====================================")
print("PyStart Katılımcı Kartı")
print("====================================")
print("Ad Soyad     :", ad_soyad)
print("Bölüm        :", bolum.upper())
print(f"Doğum Tarihi : {dogum[0]:02d}.{dogum[1]:02d}.{dogum[2]}")
print("Yaş          :", yas)
print("====================================")
```

Soru 2: Cümle Analizi
Kullanıcıdan bir cümle al ve şunları yazdır:

1 Karakter sayısı ve boşluklar hariç karakter sayısı
2 Kelime sayısı, ilk kelime ve son kelime
3 Cümlenin tamamen büyük harfli hali ve tersten yazılışı
4 Cümlede kaç tane 'a' harfi olduğu (büyük-küçük fark etmeden)
5 Cümlede 'python' kelimesinin geçip geçmediği (True ya da False)
Gözlem: İçinde 'i' harfi olan bir cümlenin büyük harfli haline dikkatle bak. Ne fark ettin? Gözlemini yorum satırı olarak yaz.

```python
# --- Soru 2: Cümle Analizi ---

metin = input("Bir cumle yaziniz: ")

# 1. Karakter sayıları
toplam_karakter = len(metin)
bosluksuz_metin = metin.replace(" ", "")
bosluksuz_karakter = len(bosluksuz_metin)

print("Karakter sayisi:", toplam_karakter)
print("Bosluklar haric karakter sayisi:", bosluksuz_karakter)

# 2. Kelime bilgileri
parcalar = metin.split()
print("Kelime sayisi:", len(parcalar))
print("Ilk kelime:", parcalar[0])
print("Son kelime:", parcalar[-1])

# 3. Buyuk harf ve tersten yazilis
print("Tamamen buyuk harfli:", metin.upper())
print("Tersten yazilisi:", metin[::-1])

# 4. 'a' harfi sayisi (kucuk harfe cevirip saydiriyoruz)
a_sayisi = metin.lower().count("a")
print("'a' harfi sayisi:", a_sayisi)

# 5. 'python' kelimesi kontrolu
python_var_mi = "python" in metin.lower()
print("'python' geciyor mu?:", python_var_mi)

# Gozlem:
# Python varsayilan olarak Ingilizce karakter kumesi kullandigi icin
# kucuk 'i' harfini buyuturken Turkce'deki 'İ' yerine Ingilizce buyuk 'I' harfine donusturur.
```

Soru 3: Not Listesi
Kullanıcıdan üç notu tek satırda, aralarında boşlukla al (70 85 90). Notları sayıya çevirip bir listeye koy, ardından:

1 Kullanıcıdan bir not daha alıp listeye ekle.
2 Listeyi küçükten büyüğe sırala ve yazdır.
3 En düşük notu, en yüksek notu ve ortalamayı yazdır.
4 En düşük notu listeden çıkar ve yeni ortalamayı yazdır.
İpucu: split() metni boşluklardan böler ama parçalar hâlâ metindir.

Örnek Çıktı:

3 notu aralarında boşluk bırakarak yaz: 70 85 90
Bir not daha ekle: 60
Notlar (sıralı): [60, 70, 85, 90]
En düşük: 60 | En yüksek: 90
Ortalama: 76.25
En düşük not çıkarıldı: [70, 85, 90]
Yeni ortalama: 81.67

```python
# --- Soru 3: Not Listesi ---

# Kullanicidan 3 notu alip parcalara boluyoruz
girilen_notlar = input("3 notu aralarinda bosluk birakarak yaz: ").split()

# Metin olarak gelen notlari sayiya cevirip tek seferde listeye koyuyoruz
puanlar = [int(girilen_notlar[0]), int(girilen_notlar[1]), int(girilen_notlar[2])]

# 1. Kullanicidan bir not daha alip listeye ekleme
ekstra_not = int(input("Bir not daha ekle: "))
puanlar.append(ekstra_not)

# 2. Listeyi kucukten buyuge siralama
puanlar.sort()
print(f"Notlar (sıralı): {puanlar}")

# 3. En dusuk, en yuksek ve ortalama
en_kucuk = min(puanlar)
en_buyuk = max(puanlar)
ortalama = sum(puanlar) / len(puanlar)

print(f"En düşük: {en_kucuk} | En yüksek: {en_buyuk}")
print(f"Ortalama: {ortalama:.2f}")

# 4. En dusuk notu cikarip yeni ortalamayi hesaplama
puanlar.remove(en_kucuk)
yeni_ortalama = sum(puanlar) / len(puanlar)

print(f"En düşük not çıkarıldı: {puanlar}")
print(f"Yeni ortalama: {yeni_ortalama:.2f}")
```

Soru 4: Telefon Rehberi
Şu sözlükle başla:

rehber = {'Ali': '0555 111 22 33', 'Ayşe': '0532 444 55 66'}
1 Kullanıcıdan bir isim ve numara alıp rehbere ekle.
2 Ali'nin numarasını '0555 999 88 77' olarak güncelle.
3 Ayşe'yi rehberden sil ve silinen numarayı yazdır.
4 Kullanıcıdan bir isim al ve numarasını yazdır. İsim rehberde yoksa program hata vermesin, 'Kayıt bulunamadı' yazsın.
5 Rehberdeki kişi sayısını ve isimleri alfabetik sırayla yazdır.
İpucu: Sözlüklerin get() metoduna ikinci bir varsayılan değer verebilirsin.

```python
# --- Soru 4: Telefon Rehberi ---

# Baslangic sozrugu
rehber = {'Ali': '0555 111 22 33', 'Ayşe': '0532 444 55 66'}

# 1. Kullanicidan kisi ve numara alip ekleme
yeni_kisi = input("Rehbere eklenecek kisi: ")
yeni_tel = input("Telefon numarasi: ")
rehber[yeni_kisi] = yeni_tel

# 2. Ali'nin numarasini guncelleme
rehber['Ali'] = '0555 999 88 77'

# 3. Ayse'yi rehberden silme ve silinen numarayi gosterme
silinen_tel = rehber.pop('Ayşe')
print("Silinen numara:", silinen_tel)

# 4. Numara sorgulama (get fonksiyonu ile hata vermeden kontrol)
sorgu = input("Numarasini gormek istediginiz kisi: ")
print(rehber.get(sorgu, "Kayıt bulunamadı"))

# 5. Kisi sayisi ve alfabetik isim listesi
print("Toplam kisi sayisi:", len(rehber))
print("Alfabetik liste:", sorted(rehber.keys()))
```

Soru 5: Ortak Hobiler
İki kişinin hobilerini virgülle ayrılmış olarak al (futbol,satranç,müzik) ve her birini bir kümeye (set) çevir. Şunları yazdır:

1 Ortak hobiler, yalnızca birinci kişide olanlar, yalnızca ikinci kişide olanlar ve tüm hobiler
2 Şu soruların cevabı (True ya da False olarak):
Ortak hobiniz var mı?
Hobi sayılarınız eşit mi?
İkiniz de en az 3 hobi yazdınız mı?
'satranç' ikinizden birinin hobisi mi?
İpucu: Kümeleri yazdırırken sıra her çalıştırmada değişebilir; sorted() ile sabitleyebilirsin.

```python
# --- Soru 5: Ortak Hobiler ---

# Hobileri virgülle ayırarak alıp kümeye (set) çeviriyoruz
hobi1_girdi = input("1. kişinin hobilerini virgülle yazınız: ")
hobi2_girdi = input("2. kişinin hobilerini virgülle yazınız: ")

# split(",") ile virgülden bölüp set() ile küme yapıyoruz
kisi1 = set(hobi1_girdi.split(","))
kisi2 = set(hobi2_girdi.split(","))

# 1. Küme İşlemleri
ortaklar = kisi1.intersection(kisi2)       # İkisinde de olanlar
sadece_birinci = kisi1.difference(kisi2)   # Yalnızca ilk kişide olanlar
sadece_ikinci = kisi2.difference(kisi1)    # Yalnızca ikinci kişide olanlar
hepsi = kisi1.union(kisi2)                 # Tüm hobiler (birleşim)

print("Ortak hobiler:", sorted(ortaklar))
print("Yalnızca birinci kişinin hobileri:", sorted(sadece_birinci))
print("Yalnızca ikinci kişinin hobileri:", sorted(sadece_ikinci))
print("Tüm hobiler:", sorted(hepsi))

# 2. Soruların Cevapları (True / False)
print("Ortak hobiniz var mı?", len(ortaklar) > 0)
print("Hobi sayılarınız eşit mi?", len(kisi1) == len(kisi2))
print("İkiniz de en az 3 hobi yazdınız mı?", len(kisi1) >= 3 and len(kisi2) >= 3)
print("'satranç' ikinizden birinin hobisi mi?", "satranç" in hepsi)
```
