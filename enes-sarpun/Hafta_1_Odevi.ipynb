### Soru 1: Katılımcı Kartı

Kullanıcıdan adını, soyadını, bölümünü ve doğum tarihini (gün, ay ve yıl ayrı ayrı) `input()` ile al, aşağıdaki gibi bir kart yazdır.

- Doğum tarihini `(gün, ay, yıl)` şeklinde bir **tuple**'da tut.
- Yaşı `2026 - doğum yılı` olarak hesapla.
- Ad soyadın baş harflerini büyüt, tarihi iki haneli yazdır (`07.03.2006`).

> **İpucu:** `f"{gun:02d}"` sayıyı iki haneye tamamlar.

**Örnek Çıktı Biçimi:**
```text
========================================
PyStart Katılımcı Kartı
========================================
Ad Soyad    : Ayşe Yılmaz
Bölüm       : Bilgisayar Mühendisliği
Doğum       : 07.03.2006
Yaş         : 20
========================================
```

````python
# --- Soru 1 Kod Alanı ---
# Kodunuzu buraya yazınız.
ad=input("Adınızı giriniz: ")
soyad=input("Soyadınızı giriniz : ")
bolum=input("Bölümünüzü giriniz:")
gun=int(input("Doğum gününüzü giriniz:"))
ay=int(input("Doğum ayınızı giriniz:"))
yil=int(input("Doğum yılınızı giriniz:"))
dogum_tarihi=(gun,ay,yil)
yas=2026-dogum_tarihi[2]
tam_ad=f"{ad} {soyad}".title()
bolum_formatli=bolum.title()
tarih_str=f"{gun:02d}.{ay:02d}.{yil}"
print("\nPyStart Katılımcı Kartı ")
print("--------------------------")
print (f"Ad Soyad :{tam_ad}")
print(f"Bölüm :{bolum_formatli}")
print(f"Doğum :{tarih_str}")
print(f"Yaş :{yas}")

````

````python
# --- Soru 3 Kod Alanı ---
# Kodunuzu buraya yazınız.
notlar_giris=input("3 notu aralarında boşluk bırakarak yaz :")
notlar_parcalari=notlar_giris.split()
notlar=[]
for not_str in notlar_parcalari:
    notlar.append(int(not_str))
yeni_not=int(input("Bir not daha ekle :"))
notlar.append(yeni_not)
notlar.sort()
print("Notlar (sıralı):",notlar)
en_dusuk=min(notlar)
en_yuksek=max(notlar)
ortalama=sum(notlar)/len(notlar)
print("En düşük:",en_dusuk," |En yüksek:",en_yuksek)
print("Ortalama:",ortalama)
notlar.remove(en_dusuk)
print("En düşük not çıkarıldı:",notlar)
print("Yeni ortalama:",round (ortalama,2))
    



````

### Soru 5: Ortak Hobiler

İki kişinin hobilerini virgülle ayrılmış olarak al (`futbol,satranç,müzik`) ve her birini bir kümeye (`set`) çevir. Şunları yazdır:

1. Ortak hobiler, yalnızca birinci kişide olanlar, yalnızca ikinci kişide olanlar ve tüm hobiler
2. Şu soruların cevabı (`True` ya da `False` olarak):
   - Ortak hobiniz var mı?
   - Hobi sayılarınız eşit mi?
   - İkiniz de en az 3 hobi yazdınız mı?
   - 'satranç' ikinizden birinin hobisi mi?

> **İpucu:** Kümeleri yazdırırken sıra her çalıştırmada değişebilir; `sorted()` ile sabitleyebilirsin.

````python
# --- Soru 5 Kod Alanı ---
# Kodunuzu buraya yazınız.
girdi1=input("1.Kişinin hobilerini giriniz:")
girdi2=input("2. Kişinin hobilerini giriniz:")
kume1=set(girdi1.split(","))
kume2=set(girdi2.split(","))
ortak=kume1&kume2
yalniz_1=kume1-kume2
yalniz_2=kume2-kume1
tum_hobiler=kume1|kume2
print("Ortak hobiler:",sorted(ortak))
print("Yalnızca ikinci kişide olanlar:",sorted(yalniz_1))
print("Yalnızca ikinci kişide olanlar :",sorted(yalniz_2))
print("Tüm hobiler :",sorted(tum_hobiler))
print("-----------------")
print("Ortak hobiniz var mı?:,len(ortak)>0")
print("Hobi sayılarınız eşit mi?:",len(kume1)==len(kume2))
print("İkiniz de en az 3 hobi yazdınız mı?:",len(kume1)>=3 and len(kume2)>=3)
print("'satranç'ikinizden birini  hobisi mi?:","satranç"in tum_hobiler)

````
