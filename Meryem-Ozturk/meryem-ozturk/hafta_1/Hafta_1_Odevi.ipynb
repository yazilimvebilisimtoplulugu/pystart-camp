### Soru 1: Katılımcı Kartı

Kullanıcıdan adını, soyadını, bölümünü ve doğum tarihini (gün, ay ve yıl ayrı ayrı) `input()` ile al, aşağıdaki gibi bir kart yazdır.

- Doğum tarihini `(gün, ay, yıl)` şeklinde bir **tuple**'da tut.
- Yaşı `2026 - doğum yılı` olarak hesapla.
- Ad soyadın baş harflerini büyüt, tarihi iki haneli yazdır (`07.03.2006`).

> **İpucu:** `f"{gun:02d}"` sayıyı iki haneye tamamlar.

**Örnek Çıktı Biçimi:**
```text
========================================
PyStart Katılımcı Kartı
========================================
Ad Soyad    : Ayşe Yılmaz
Bölüm       : Bilgisayar Mühendisliği
Doğum       : 07.03.2006
Yaş         : 20
========================================
```

````python
# --- Soru 1 Kod Alanı ---
# Soru:1 Katılımcı Kartı

#ilk olarak degiskenleri input ile alıyoruz.

ad = input("Lütfen adınızı giriniz: ")
soyad = input("Lütfen soyadınızı giriniz: ")
bolum = input("Lütfen bölümünüzü giriniz: ")
dogum_yili = int(input("Lütfen doğum yılınızı giriniz: "))
dogum_ayi = int(input("Lütfen doğum ayınızı giriniz: "))
dogum_gunu = int(input("Lütfen doğum gününüzü giriniz: "))

#dogum tarihini bir tuple olarak saklıyoruz.

dogum_tarihi =(dogum_gunu, dogum_ayi, dogum_yili)

yas= 2026 - dogum_yili  #yas hesaplaması yapıyoruz. 
#Bunu ayrıca yas= 2026 - dogum_tarihi[2] şeklinde de yapabiliriz.

ad_soyad = f"{ad.title()} {soyad.title()}"  #ad ve soyadı birleştiriyoruz ve title() ile baş harflerini büyük yapıyoruz.

dogum_tarihi_yazimi = f"{dogum_tarihi[0]:02d}/{dogum_tarihi[1]:02d}/{dogum_tarihi[2]}"  #dogum tarihini formatlıyoruz.

print("========================================")
print("PyStart Katılımcı Kartı")
print("========================================")

print(f"Ad Soyad: {ad_soyad}")
print(f"Bölüm: {bolum.title()}") #title() aynı zamanda burada da kullanılabilir.
print(f"Doğum Tarihi: {dogum_tarihi_yazimi}")
print(f"Yaş: {yas}")
print("========================================")

````

### Soru 2: Cümle Analizi

Kullanıcıdan bir cümle al ve şunları yazdır:

1. Karakter sayısı ve boşluklar hariç karakter sayısı
2. Kelime sayısı, ilk kelime ve son kelime
3. Cümlenin tamamen büyük harfli hali ve tersten yazılışı
4. Cümlede kaç tane 'a' harfi olduğu (büyük-küçük fark etmeden)
5. Cümlede 'python' kelimesinin geçip geçmediği (`True` ya da `False`)

> **Gözlem:** İçinde 'i' harfi olan bir cümlenin büyük harfli haline dikkatle bak. Ne fark ettin? Gözlemini yorum satırı olarak yaz.

````python
# --- Soru 2 Kod Alanı ---
#Soru 2: Cümle Analizi

cumle = input("Lütfen bir cümle giriniz: ")

#1. Karakter sayısı ve boşluklar hariç karakter sayısı

karakter_sayisi = len(cumle)  #cümledeki karakter sayısını hesaplıyoruz.

print(f"Cümledeki karakter sayısı: {karakter_sayisi}")

bosluksuz_karakter_sayisi = len(cumle.replace(" ", ""))  #cümledeki boşlukları kaldırıp karakter sayısını hesaplıyoruz.

print(f"Cümledeki boşluksuz karakter sayısı: {bosluksuz_karakter_sayisi}")

#2. Kelime sayısı, ilk kelime ve son kelime

kelimeler = cumle.split()  #cümleyi kelimelere ayırıyoruz.
kelime_sayisi = len(kelimeler)
print(f"Cümledeki kelime sayısı: {kelime_sayisi}")
print(f"Cümlenin ilk kelimesi: {kelimeler[0]}") #python split() ile ayırdığı kelimleri liste yaptığı için ilk kelimeye 0. index ile ulaşabiliriz.
print(f"Cümlenin son kelimesi: {kelimeler[-1]}") #son kelimeye -1 index ile ulaşabiliriz.


#3. Cümlenin tamamen büyük harfli hali ve tersten yazılışı

print(cumle.upper())  #cümleyi tamamen büyük harfli yapıyoruz.

print(cumle[::-1])  #cümleyi tersten yazdırıyoruz.

#4. Cümlede kaç tane 'a' harfi olduğu (küçük-büyük fark etmeden)

print(f"Cümledeki 'a' harfi sayısı: {cumle.lower().count('a')}")  #cümlede büyük küçük farketmemesi için tüm cümleyi küçük harfe çevirdik ve count() ile de sayısını aldık.

#5. Cümlede 'python' kelimesinin geçip geçmediği

print("python" in cumle.lower()) #Soru 4 deki gibi python büyük küçük yazılmasının farketmemesi için cümleyi küçük harfe çevirdik ve in ile python kelimesinin geçip geçmediğini kontrol ettik.

# Gözlem Yorumunuz:Küçük i harfi pythonda büyük harfe döndürüldüğünde I harfine dönüşüyor.
````

### Soru 3: Not Listesi

Kullanıcıdan üç notu tek satırda, aralarında boşlukla al (`70 85 90`). Notları sayıya çevirip bir listeye koy, ardından:

1. Kullanıcıdan bir not daha alıp listeye ekle.
2. Listeyi küçükten büyüğe sırala ve yazdır.
3. En düşük notu, en yüksek notu ve ortalamayı yazdır.
4. En düşük notu listeden çıkar ve yeni ortalamayı yazdır.

> **İpucu:** `split()` metni boşluklardan böler ama parçalar hâlâ metindir.

**Örnek Çıktı:**
```text
3 notu aralarında boşluk bırakarak yaz: 70 85 90
Bir not daha ekle: 60
Notlar (sıralı): [60, 70, 85, 90]
En düşük: 60 | En yüksek: 90
Ortalama: 76.25
En düşük not çıkarıldı: [70, 85, 90]
Yeni ortalama: 81.67
```

````python
# --- Soru 3 Kod Alanı ---
#Soru 3: Not Listesi

notlar = input("3 notu aralarında boşluk bırakarak yaz: ").split() #boşluk olan sayılar string olarak alınır.

not_listesi =[]
not_listesi.append(int(notlar[0])) #hem int hem de append ile listeye ekliyoruz.
not_listesi.append(int(notlar[1]))
not_listesi.append(int(notlar[2]))

yeni_not = int(input("Bir not daha ekle: "))

not_listesi.append(yeni_not)

not_listesi.sort() #büyükten küçüğe sıralamak için sort() kullanıyoruz.

print(f"Notlar (sıralı): {not_listesi}") 

en_dusuk = min(not_listesi)
en_yuksek = max(not_listesi)

print(f"En düşük: {en_dusuk} | En yüksek: {en_yuksek}")

not_listesi = [60, 70, 85, 90]

ortalama = sum(not_listesi) / len(not_listesi)

print(f"Ortalama: {ortalama:.2f}")

not_listesi.remove(en_dusuk)

print(f"En düşük not çıkarıldı: {not_listesi}")

ortalama = sum(not_listesi) / len(not_listesi)

print(f"Ortalama: {ortalama:.2f}")

````

### Soru 4: Telefon Rehberi

Şu sözlükle başla:
```python
rehber = {'Ali': '0555 111 22 33', 'Ayşe': '0532 444 55 66'}
```

1. Kullanıcıdan bir isim ve numara alıp rehbere ekle.
2. Ali'nin numarasını `'0555 999 88 77'` olarak güncelle.
3. Ayşe'yi rehberden sil ve silinen numarayı yazdır.
4. Kullanıcıdan bir isim al ve numarasını yazdır. İsim rehberde yoksa program hata vermesin, `'Kayıt bulunamadı'` yazsın.
5. Rehberdeki kişi sayısını ve isimleri alfabetik sırayla yazdır.

> **İpucu:** Sözlüklerin `get()` metoduna ikinci bir varsayılan değer verebilirsin.

````python
rehber = {'Ali': '0555 111 22 33', 'Ayşe': '0532 444 55 66'}

# --- Soru 4 Kod Alanı ---

isim = input("Eklemek istediğiniz kişinin ismini giriniz: ")
numara= input("Telefon numarası: ")
rehber[isim] = numara #yeni kullanıcı ekledik.

rehber['Ali'] = '0555 999 88 77' #Ali'nin numarasını güncelledik.

silinen_numara = rehber['Ayşe'] #silinen numarayı önce kaydetmeliyiz ki silindikten sonra ekrana yazdırabilelim.

rehber.pop('Ayşe')

print("Silinen numara:", silinen_numara)

aranan_isim = input("Numarasını öğrenmek istediğiniz kişinin adı: ")

numara = rehber.get(aranan_isim, "Kayıt bulunamadı")

print(numara)

print("Rehberdeki kişi sayısı:", len(rehber))

print(sorted(rehber.keys()))

````

### Soru 5: Ortak Hobiler

İki kişinin hobilerini virgülle ayrılmış olarak al (`futbol,satranç,müzik`) ve her birini bir kümeye (`set`) çevir. Şunları yazdır:

1. Ortak hobiler, yalnızca birinci kişide olanlar, yalnızca ikinci kişide olanlar ve tüm hobiler
2. Şu soruların cevabı (`True` ya da `False` olarak):
   - Ortak hobiniz var mı?
   - Hobi sayılarınız eşit mi?
   - İkiniz de en az 3 hobi yazdınız mı?
   - 'satranç' ikinizden birinin hobisi mi?

> **İpucu:** Kümeleri yazdırırken sıra her çalıştırmada değişebilir; `sorted()` ile sabitleyebilirsin.

````python
# --- Soru 5 Kod Alanı ---
#Soru 5: Ortak Hobiler

kisi1 = input("1. kişinin hobileri (virgülle ayırarak yazınız): ")
kisi2 = input("2. kişinin hobileri (virgülle ayırarak yazınız): ")

kume1 = set(kisi1.split(","))
kume2 = set(kisi2.split(",")) #set() fonksiyonu ile kümeler oluşturuyoruz.

ortak = kume1 & kume2
yalniz_kisi1 = kume1 - kume2
yalniz_kisi2 = kume2 - kume1
tum_hobiler = kume1 | kume2

print("Ortak hobiler:", sorted(ortak)) #sorted kullanmak zorunda değiliz ama daha düzenli olsun diye yazdık.
print("Yalnızca birinci kişinin hobileri:", sorted(yalniz_kisi1))
print("Yalnızca ikinci kişinin hobileri:", sorted(yalniz_kisi2))
print("Tüm hobiler:", sorted(tum_hobiler))

print("Ortak hobiniz var mı?", bool(ortak))
print("Hobi sayılarınız eşit mi?", len(kume1) == len(kume2))
print("İkiniz de en az 3 hobi yazdınız mı?", len(kume1) >= 3 and len(kume2) >= 3)
print("'satranç' ikinizden birinin hobisi mi?", "satranç" in kume1 or "satranç" in kume2)

````
